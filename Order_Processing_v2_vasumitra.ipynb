# Warehouse Batch Order Processing System (Extended with Lists)

This extends the original single-order Warehouse Order Processing System to handle a **batch of orders** collected in a list, generate an **end-of-day report**.

## Collect a batch of orders into a list

Ask how many orders to enter, then loop that many times collecting each order's details and appending them as a list to `all_orders`.

```python
all_orders = []  # will hold one list per order:
# [customer_id, order_value, num_items, distance_km, is_premium, is_priority, is_fragile]

num_orders = int(input("How many orders do you want to enter? ").strip())

for order_num in range(num_orders):
    print()
    print(f"--- Enter details for Order {order_num + 1} ---")

    customer_id = input("Enter Customer ID: ").strip()
    order_value = float(input("Enter Order value (Rs.): ").strip())
    num_items = int(input("Enter Number of items: ").strip())
    distance_km = float(input("Enter Delivery distance (in km): ").strip())

    premium_input = input("Is the customer a premium member? (Yes/No): ").strip().lower()
    is_premium = (premium_input == "yes")

    priority_input = input("Does the order require priority processing? (Yes/No): ").strip().lower()
    is_priority = (priority_input == "yes")

    fragile_input = input("Does the order contain a fragile item? (Yes/No): ").strip().lower()
    is_fragile = (fragile_input == "yes")

    # Store this order as a list, and append it to the master list
    single_order = [customer_id, order_value, num_items, distance_km, is_premium, is_priority, is_fragile]
    all_orders.append(single_order)

print()
print("All", len(all_orders), "orders collected successfully!")
```

## Process every order in the batch

Loop through `all_orders`, apply the same business rules as the original system to each one, and print an individual summary. Results are also saved into parallel lists so Step 3 can build the report.

```python
base_charge = 50
per_km_charge = 8

# Lists to hold results for each order, used later for the report
order_statuses = []             # "ACCEPTED" / "REJECTED" per order
order_shipping_costs = []       # numeric shipping cost per order (0 if free/rejected)
order_special_handling = []     # True/False per order
order_priority_processing = []  # True/False per order

separator = ""
for counter in range(35):
    separator = separator + "="

for order_index in range(len(all_orders)):
    order = all_orders[order_index]

    customer_id = order[0]
    order_value = order[1]
    num_items = order[2]
    distance_km = order[3]
    is_premium = order[4]
    is_priority = order[5]
    is_fragile = order[6]

    shipping_cost = base_charge + (per_km_charge * distance_km)

    order_accepted = (order_value > 500) and (num_items >= 1) and (distance_km <= 100)
    free_shipping = (order_value >= 2000) or (is_premium)
    priority_processing = (is_priority) and (order_value >= 1500)
    special_handling = (is_fragile) or (num_items >= 10)

    if order_accepted:
        status_label = "ACCEPTED"
    else:
        status_label = "REJECTED"

    if not order_accepted:
        shipping_label = "NOT APPLICABLE"
        processing_label = "NOT APPLICABLE"
        handling_label = "NOT APPLICABLE"
        shipping_cost_for_report = 0
    else:
        if free_shipping:
            shipping_label = "FREE"
            shipping_cost_for_report = 0
        else:
            shipping_label = "Rs." + str(round(shipping_cost, 2))
            shipping_cost_for_report = shipping_cost

        if priority_processing:
            processing_label = "PRIORITY"
        else:
            processing_label = "STANDARD"

        if special_handling:
            handling_label = "SPECIAL"
        else:
            handling_label = "NORMAL"

    # Save results for the end-of-day report
    order_statuses.append(status_label)
    order_shipping_costs.append(shipping_cost_for_report)
    order_special_handling.append(special_handling if order_accepted else False)
    order_priority_processing.append(priority_processing if order_accepted else False)

    # Print individual order summary
    if order_value == int(order_value):
        order_value_display = str(int(order_value))
    else:
        order_value_display = str(order_value)

    if distance_km == int(distance_km):
        distance_display = str(int(distance_km))
    else:
        distance_display = str(distance_km)

    print()
    print(separator)
    print("   ORDER PROCESSING RESULT - Order", order_index + 1)
    print(separator)
    print("Customer          :", customer_id)
    print("Order value       : Rs." + order_value_display)
    print("Items             :", num_items)
    print("Distance          :", distance_display, "km")
    print()
    print("Status            :", status_label)
    print("Shipping          :", shipping_label)
    print("Processing        :", processing_label)
    print("Handling          :", handling_label)
    print(separator)
```

## End-of-day report

Using only the data collected in `all_orders` and the parallel result lists, compute totals across the whole batch.

```python
total_orders = len(all_orders)
accepted_count = 0
rejected_count = 0
total_revenue = 0
total_shipping_collected = 0
special_handling_count = 0
priority_count = 0

highest_value = -1
highest_value_customer = None

for i in range(total_orders):
    order = all_orders[i]
    customer_id = order[0]
    order_value = order[1]

    if order_statuses[i] == "ACCEPTED":
        accepted_count += 1
        total_revenue += order_value
        total_shipping_collected += order_shipping_costs[i]
    else:
        rejected_count += 1

    if order_special_handling[i]:
        special_handling_count += 1

    if order_priority_processing[i]:
        priority_count += 1

    if order_value > highest_value:
        highest_value = order_value
        highest_value_customer = customer_id

print()
print(separator)
print("        END-OF-DAY REPORT")
print(separator)
print("Total orders received      :", total_orders)
print("Orders ACCEPTED             :", accepted_count)
print("Orders REJECTED             :", rejected_count)
print("Total revenue (accepted)   : Rs.", round(total_revenue, 2))
print("Total shipping collected   : Rs.", round(total_shipping_collected, 2))
print("Orders needing special handling:", special_handling_count)
print("Orders with priority processing:", priority_count)

if highest_value_customer is not None:
    print("Highest value order        :", highest_value_customer, "- Rs.", highest_value)
print(separator)
```

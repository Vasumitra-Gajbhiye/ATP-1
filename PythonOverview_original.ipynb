### Create Complex Number

````python
# Complex Number
# Method 1: Using the 'j' literal directly
num1 = 3 + 4j

# Method 2: Using the built-in complex(real, imaginary) function
num2 = complex(2, -1)  # Evaluates to 2 - 1j

print(num1)
print(type(num1))
````

#### Access the Parts complex number

````python
# --- Accessing Parts ---
# You can extract parts using the .real and .imag attributes (returned as floats)
print(num1.real)
print(num1.imag)
````

````python
# --- Arithmetic Operations ---
# You can add, subtract, multiply, and divide complex numbers directly
add_result = num1 + num2
print(add_result)
````

# Activity: The Variable Swap
You have two variables:

- cup_a = "Coffee"
- cup_b = "Tea"

Your mission: make cup_a hold "Tea" and cup_b hold "Coffee" — You can only reassign variables.

````python
# Using temporary variable

cup_a = "Coffee"
cup_b = "Tea"

print("cup_a:", cup_a)
print("cup_b:", cup_b)

temp = cup_a
cup_a = cup_b
cup_b = temp

print("cup_a:", cup_a)
print("cup_b:", cup_b)
````

````python
#no third variable needed!
cup_a = "Coffee"
cup_b = "Tea"

cup_a, cup_b = cup_b, cup_a
````

##Identifiers

A Python identifier is a name used to identify a variable, function, class, module or other object.

Rules for setting Identifiers

- can only start with an alphabet or _
- Followed by 0 or more letter, _ and digits
- keywords cannot be used as an identifiers

### Variable Naming Rules

- Must start with a letter or underscore (not a number)
- Can contain letters, numbers, and underscores only — no spaces or symbols
- Case-sensitive: `age` and `Age` are different variables
- Cannot be a Python keyword (like `if`, `for`, `True`, `print`)
- Convention: use lowercase with underscores for readability, e.g. `first_name`


````python
# Examples of VALID names
student_name = "Mike"
score_1 = 88
_temp = 20

print(student_name, score_1, _temp)
````

````python
# Examples of INVALID names
1st_place = "gold"
````

## Expressions & Operators

An **expression** is a piece of code that produces a value. Operators combine values into expressions.

````python
a = 17
b = 5

print(a + b)
````

#Operators in Python
- Arithmetic: +, -, *, /, %, **, //
- Assignment: =, +=, -=, *=, /=, %=, **=, //=
- Comparison: ==, !=, >, <, >=, <=
- Logical: and, or, not
- Bitwise: &, |, ^, ~, <<, >>
- Identity: is, is not
- Membership: in, not in

### Arithmetic operators

| Operator | Meaning | Example |
|---|---|---|
| `+` | addition | `5 + 2` → `7` |
| `-` | subtraction | `5 - 2` → `3` |
| `*` | multiplication | `5 * 2` → `10` |
| `/` | division (always gives a float) | `5 / 2` → `2.5` |
| `//` | floor (integer) division | `5 // 2` → `2` |
| `%` | modulus (remainder) | `5 % 2` → `1` |
| `**` | exponent (power) | `5 ** 2` → `25` |

### Assignment operators

These update a variable's value based on its current value — a useful shortcut.

| Operator | Same as |
|---|---|
| `x += 5` | `x = x + 5` |
| `x -= 5` | `x = x - 5` |
| `x *= 5` | `x = x * 5` |
| `x /= 5` | `x = x / 5` |


````python
score = 10
print(score)

score += 5
print(score)

score *= 2
print(score)

score -= 3
print(score)
````

## Common Beginner Errors: Syntax vs Runtime

- **Syntax Error**: your code breaks Python's grammar rules. Python won't even start running the code. (e.g., missing colon, mismatched brackets, misspelled keyword)
- **Runtime Error**: your code is grammatically valid, but something goes wrong *while it's running*. (e.g., dividing by zero, using a variable that doesn't exist, converting text that isn't a number)


````python
# Example of a SYNTAX error — Python can't even understand this
print "what is SYNTAX error")
````

````python
# Example of a RUNTIME error — this is valid Python, but fails while running
number = 10
result = number / 0
print(result)
````

## Tuples

A **tuple** is an ordered collection (preserved insertion order), written with parentheses `()`. It looks a lot like a list, but with one huge difference: **tuples are immutable** — once created, they cannot be changed.
Also The individual elements can be of any type, and need not be of the same type.

```python
point = (3, 4)
```

````python
data_points = (3, 4)
print(data_points)
print(type(data_points))
print(data_points[0], data_points[1])   # indexing works just like lists

colors = ("red", "green", "blue")
for c in colors:
    print(c)
````

### Changing Tuples to Lists

We can change tuples to lists and lists to tuples. Tuple is immutable if we want to modify a tuple we should change it to a list.

````python
fruits = ('Orange', 'Apple', 'Guava','Banana')
lst_fruits = list(fruits)
print(lst_fruits)

lst_fruits.append('Grapes')
print(lst_fruits)

fruits = tuple(lst_fruits)
print(fruits)
````

### Joining Tuples
We can join two or more tuples using + operator

````python
fruits = ('banana', 'orange', 'mango', 'lemon')
vegetables = ('Tomato', 'Potato', 'Cabbage','Onion', 'Carrot')
fruits_and_vegs = fruits + vegetables

print(fruits_and_vegs)
````

### Deleting Tuples
It is not possible to remove a single item in a tuple but it is possible to delete the tuple itself using del.

````python
fruits = ('banana', 'orange', 'mango', 'lemon')
del fruits

print(fruits)
````

# zipping tuples

This code pairs up elements from two separate tuples based on their positions and combines them into a single tuple of pairs.

````python
a = (1, 2, 3, 4)
b = (5, 6, 7, 8)
print(tuple(zip(a, b)))

a = (1, 2, 3, 4, 99, 100)  # 6 elements
b = (5, 6, 7, 8)           # 4 elements

print(tuple(zip(a, b))) # 99 and 100 are left out because b ran out of items!
````

## Dictionaries — Key-Value Pairs

A **dictionary** stores data as **key-value pairs**, written with curly braces `{}`. Instead of looking things up by position (like a list), you look them up by a meaningful **key**.

```python
student = {"name": "Bob", "age": 17, "grade": "11th"}
```

````python
student = {"name": "Bob", "age": 17, "grade": "11th"}

print(student)
print(student["name"])     # access by key
print(student["age"])

# Check what happens if key is missing while accessing
print(student["score"])
````

````python
# .get() is a safer way to access — returns None (or a default) instead of an error
# if the key is missing

print(student.get("grade"))
print(student.get("school"))
print(student.get("school", "Not specified"))
````

### Key Rules About Dictionaries
- No Indexing: No index support.
- Mutable: Can modify post-creation.
- Keys: Immutable (str, num, tuple); Values: Any type.
- Unique Keys: Keys must be unique; overwrite on reassignment.

````python
example_dictionary = {"Name":"Saurabh", "Gender":"Male"}
print(example_dictionary)
````

````python
example_dictionary = {[1, 2, 3]:"Saurabh"}
print(example_dictionary)
````

````python
# with mixed keys
example_dictionary = {(1, 2, 3):"Jonny"}
example_dictionary

# Option A: With parentheses
print(example_dictionary[(1, 2, 3)])

# Option B: Without parentheses (Python implicitly treats it as a tuple)
print(example_dictionary[1, 2, 3])
````

````python
# duplicate keys
example_dictionary = {"Name":"Nate", "Name":"Paul"}
example_dictionary
````

#### The value in dictionary could be any data types:string, boolean, list, tuple, set or a dictionary.

````python
old_person = {
    'first_name':'Dadaji',
    'last_name':'Rao',
    'age':250,
    'country':'India',
    'is_married':True,
    'skills':['JavaScript', 'C', 'C++', 'R'],
    'address':{
        'street':'Bakshi street',
        'pincode':'560420'
    }
}

print(old_person)
print(old_person['skills'])
print(old_person['skills'][0])
print(old_person['address']['pincode'])
print(old_person.get("address").get("street"))

# Then the second .get() safely looks inside {} and returns "Not Found"
print(old_person.get('address', {}).get('city', 'Not Found'))
print(old_person.get('ship_address', {}).get('street', 'Not Found'))

#Since skills is a list we can use Append function
old_person['skills'].append('HTML')
print(old_person['skills'])
````

### Adding, updating, and deleting entries


````python
student = {"name": "Mark", "age": 17}
print("Before:", student)

student["grade"] = "11th"       # adds a new key-value pair
print("After adding grade:", student)

student["age"] = 18             # updates an existing key
print("After updating age:", student)

del student["grade"]            # deletes a key-value pair
print("After deleting grade:", student)

# .pop() also deletes, and returns the removed value
#pop(key): removes the item with the specified key name:
#popitem(): removes the last item
removed_age = student.pop("age")
print("Removed age:", removed_age)
print("dict post removal of age:", student)

student.popitem()
print("Final dict:", student)
````

#### Clearing a Dictionary
If we don't want the items in a dictionary we can clear them using clear() method

````python
student = {"name": "Mark", "age": 17}
print(student.clear())
````

#### Copy a Dictionary
We can copy a dictionary using a copy() method. Using copy we can avoid mutation of the original dictionary.

````python
student = {"name": "Mark", "age": 17}
new_student = student
new_student_copy = student.copy()

student["age"] = 18
print(student)
print(new_student)
print(new_student_copy)
````

### Looping over a dictionary


````python
student = {"name": "John", "age": 16, "grade": "10th"}

print("Keys only:")
for key in student:
    print(key)

print("\nKeys and values with .items():")
for key, value in student.items():
    print(key, "->", value)

print("\nAll keys:", list(student.keys()))
print("All values:", list(student.values()))
````

## Practice — Model Real Data: Contact Book

A dictionary is perfect for modeling real-world records, since each field has a meaningful name (key). Let's build a small **contact book** — a dictionary of dictionaries!

## Compare Tuple vs Dictionary Use Cases

| | Tuple | Dictionary |
|---|---|---|
| Written as | `(1, 2, 3)` | `{"a": 1, "b": 2}` |
| Ordered? | Yes | Yes (insertion order, Python 3.7+) |
| Mutable? | **No** | Yes |
| Accessed by | position/index | key |
| Best for | fixed data, coordinates, function returns | labeled/structured records, lookups by name |

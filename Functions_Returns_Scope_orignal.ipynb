## Local vs. Global Variables

- A variable created **inside** a function is **local** — it only exists there.
- A variable created **outside** any function is **global** — readable from anywhere, but if you want to *modify* it from inside a function, you must say `global` keyword.


````python
# Variable scope & function behavior

def incrementvalue(y):
    x = 1                 # Local x
    x += 1
    print(x)

x = 5                     # Global x
incrementvalue(x)         # Calls function
print(x)

# Functions have local scope. Global vars coexist but are not affected.
````

#### Local Variables: Inside function.
#### Global Variables: Outside any function, in main program.

## Nested Functions

A **nested function** is a function defined *inside* another function. The inner function is only accessible from within the outer function — it doesn't exist anywhere else.

```python
def outer_function():
    print("I am in outer function")
    def inner_function():
        print("I am in outer_function--> inner_function")
    inner_function()

outer_function()
inner_function()   # this would ERROR - inner_function doesn't exist out here
```


````python
def outer_function():
    print("I am in outer function")
    def inner_function():
        print("I am in outer_function--> inner_function")
    inner_function()

outer_function()
````

### Why would you ever do this?

Nested functions are useful when a helper task is ONLY needed by one specific function, and doesn't make sense as a standalone, reusable piece of code on its own. It keeps your code organized and avoids cluttering the rest of your program with a function nobody else needs.

#### The interesting part: the inner function can "see" the outer function's variables

This is called using the **enclosing scope**.


````python
def greet_user(name):
    greeting = "Hello"    # local to greet_user

    def build_message():
        return greeting + ", " + name + "!"   # can read 'greeting' and 'name' from the outer function!

    return build_message()

print(greet_user("John"))
````

### Modifying an outer (but not global) variable: `nonlocal`

If the inner function needs to **change** a variable from the enclosing function (not a fully global one), you need the `nonlocal` keyword — the enclosing-scope equivalent of `global`.


````python
def make_counter():
    count = 0

    def increment():
        nonlocal count     # without this, Python would treat 'count' as a new local variable
        count += 1
        return count

    print(increment())
    print(increment())
    print(increment())

make_counter()
````

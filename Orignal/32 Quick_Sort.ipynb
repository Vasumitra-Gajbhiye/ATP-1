# Quick Sort

```python
def partition(arr, low, high):
    # Choose the first element as the pivot
    pivot = arr[low]

    # Initialize i at the start and j at the end
    i = low
    j = high

    while i < j:
        # Move i to the right while elements are less than or equal to the pivot
        while i <= high - 1 and arr[i] <= pivot:
            i += 1

        # Move j to the left while elements are greater than the pivot
        while j >= low + 1 and arr[j] > pivot:
            j -= 1

        # If pointers haven't crossed, swap the out-of-place elements
        if i < j:
            arr[i], arr[j] = arr[j], arr[i]

    # Place the pivot in its final sorted position by swapping with j
    arr[low], arr[j] = arr[j], arr[low]
    return j
```

```python
def quick_sort(arr, low, high):
    if low < high:
        # Find the partition index where the pivot is placed
        p_index = partition(arr, low, high)

        # Recursively sort elements before and after the partition
        quick_sort(arr, low, p_index - 1)
        quick_sort(arr, p_index + 1, high)
```

```python
nums = [10, 7, 8, 9, 1, 5]
quick_sort(nums, 0, len(nums) - 1)
print("Sorted array:", nums)
```

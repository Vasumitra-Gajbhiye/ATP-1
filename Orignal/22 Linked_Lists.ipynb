# Data Structures — Linked Lists


#**What is data structure?**

A data structure is how we organize data plus the operations it allows, chosen so that the operations you need are cheap.

There are 2 types:
- Linear (Array, Linked list, Stack, Queue etc)
- Non-Linear (Trees, Graph)

# Linked List
- A linked list is a linear data structure used as an alternative to a traditional array.
- Non-Contiguous Storage: Unlike arrays, it stores data in non-contiguous (scattered) memory locations.


**How It Works (The Structure)**

- Collection of Nodes: The structure consists of a sequence of objects called nodes.
- Node Composition: Each node acts as a container that stores two things: the actual value/data and the memory address (pointer) of the next node in the sequence.
- Flexible Data Types: The data stored inside a node can be of any data type (e.g., integers, strings, or custom objects).


**Key Markers (Head & Tail)**

- The Head: The very first node in the sequence is called the head, which serves as the entry point to the list.
- The Tail: The final node in the sequence is called the tail.
- End Indicator: The tail's address pointer points to None (or null), signaling the absolute end of the data structure.

## Linked List Concept vs. Array/List, Node Structure

A Python `list` stores its items in one continuous, numbered block of memory — that's exactly why `my_list[3]` is instant: Python can jump straight to position 3 by doing simple math on the starting address.

A **linked list** stores items completely differently: each item (called a **node**) holds its **data**, plus a pointer to the **next** node in the chain. There's no single continuous block — nodes can live anywhere in memory, connected only by these pointers.

```
[10 | next] -> [20 | next] -> [30 | next] -> None
```

**Key difference to internalize:** to reach the 3rd node in a linked list, you can't jump there directly — you MUST start at the first node and follow the `next` pointers one at a time until you arrive. There is no shortcut.

### The Node — the building block of a linked list


````python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None   # starts pointing at nothing
````

# Manually linking three nodes together
```
node1 = Node(10)
node2 = Node(20)
node3 = Node(30)

node1.next = node2   # 10 now points to 20
node2.next = node3   # 20 now points to 30
node3.next stays None - it's the last node

print(node1.data, node1.next.data, node1.next.next.data)
```

## Linked List vs. Array-Based List — Trade-offs

| | Array-based list (Python `list`) | Linked list |
|---|---|---|
| **Access by position**  | Fast — direct jump, O(1) | Slow — must walk from the head, O(n) |
| **Insert/delete at the END** | Fast (usually) | Slow — must walk to the end first (unless you keep a `tail` pointer) |
| **Insert/delete at the FRONT** | Slow — every other element must shift over, O(n) | Fast — just re-point the head, O(1) |
| **Memory layout** | One continuous block | Scattered nodes, connected by pointers |
| **Extra memory overhead** | None beyond the data itself | Every node stores an extra `next` pointer |
| **Best for** | Frequent lookups by position/index | Frequent insertions/deletions at the front, or when size changes a lot |

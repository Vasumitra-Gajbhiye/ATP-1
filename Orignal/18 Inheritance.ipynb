# Inheritance

The Child class is called a derived class or subclass. The Parent class is known as base class or superclass. Parent is specified in () after the class name, class Child(Parent).

With inheritance, you are taking an existing class and:

- Adding new methods
- Redefining some of the existing methods
- Adding new attributes to instances
In the end you are extending existing code.

````python
class User: # Parent

    def login(self):
        print('login')

    def register(self):
        print('Register')
````

````python
class Student(User): # Child

    def enroll(self):
        print('Enroll')

    def review(self):
        print('Review')
````

````python
stu1 = Student()

stu1.enroll()
stu1.review()
stu1.login()
stu1.register()
````

````python
u = User()

u.login()
u.register()
````

````python
u.enroll()
u.review()
````

# Constructor

- `class B` inherits `class A`

- No constructor in `B` ---> `A`'s constructor invoked

- Parent class constructor used if child lacks one

````python
class Stock:
    def __init__(self, name, shares, price):
        self.name = name
        self.shares = shares
        self.price = price
        print ("Stock Constructor: ", self.name)

    def cost(self):
        return self.shares * self.price

    def sell(self, nshares):
        self.shares -= nshares
````

````python
class MyStock(Stock):
    pass
````

````python
s = MyStock('GOOGLE', 100, 490.1)
````

````python
s.sell(25)
````

````python
s.shares
````

# Child class will not be able to access Private Members of parent class

````python
class Stock:
    def __init__(self, name, shares, price):
        self.name = name
        self.shares = shares
        self.__price = price

    def cost(self):
        return self.shares * self.__price

    def sell(self, nshares):
        self.shares -= nshares
````

````python
class MyStock(Stock):
    pass
````

````python
s = MyStock('GOOG', 100, 400)

print(s.shares)
````

````python
print(s.__price)
````

## Method Overriding
- Child Class method overrides Parent Class method if they have the same name.

````python
class Stock:
    def __init__(self, name, shares, price):
        self.name = name
        self.shares = shares
        self.price = price

    def cost(self):
        return self.shares * self.price

    def sell(self, nshares):
        self.shares -= nshares
````

````python
class MyStock(Stock):
    def cost(self):
        return 1.25 * self.shares * self.price
````

````python
s = MyStock('GOOG', 100, 400)

print(s.cost())
````

# You can add new methods in child class

````python
class Stock:
    def __init__(self, name, shares, price):
        self.name = name
        self.shares = shares
        self.price = price

    def cost(self):
        return self.shares * self.price

    def sell(self, nshares):
        self.shares -= nshares
````

````python
class MyStock(Stock):
    def panic(self):
        self.sell(self.shares)
````

````python
s = MyStock('GOOG', 100, 1000)
s.sell(25)
print(s.shares)
````

````python
s.panic()
print(s.shares)
````

# Overriding
Sometimes a class extends an existing method, but it wants to use the original implementation inside the redefinition. For this, use super():

````python
class Stock:
    def __init__(self, name, shares, price):
        self.name = name
        self.shares = shares
        self.price = price

    def cost(self):
        return self.shares * self.price

    def sell(self, nshares):
        self.shares -= nshares
````

````python
class MyStock(Stock):
    def cost(self):
        actual_cost = super().cost()
        print("actual cost from parent is: ", actual_cost)
        return 1.25 * actual_cost
````

````python
s = MyStock('GOOG', 100, 1000)
````

````python
s.cost()
````

# In Summary

- Child class cannot acess private data members of the the parent class.
- If Child class constructor not given, Parent class construtor is executed.
- Child class can override the method from parent class

## Super Keyword
- Accesses parent class methods
- Accesses parent class constructor
- Does NOT access parent class attributes

````python
class Phone:
    def __init__(self, price, brand, camera):
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")
````

````python
class SmartPhone(Phone):
    def buy(self):
        print ("Buying a smartphone")
        super().buy()
````

````python
s = SmartPhone(20000, "Apple", 13)
````

````python
s.buy()
````

````python
s.super().buy() # super keyword will not work outside of the class
````

# Multilevel level Inheritence

````python
class Product:
    def review(self):
        print ("Product customer review")

class Phone(Product):
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera
    def buy(self):
        print ("Buying a phone")

class SmartPhone(Phone):
    def __init__(self, price, brand, camera, os, ram):
        print ("Inside smartphone constructor")
        super().__init__(price, brand, camera)
        self.os = os
        self.ram = ram
    def buy(self):
        print ("Buying a smartphone")
````

````python
s = SmartPhone(20000, "Apple", 12, "IOS", 16)
````

````python
s.brand
````

````python
s.os
````

````python
p = Phone(1000, "Samsung", 1)
````

````python
s.review()
````

````python
p.review()
````

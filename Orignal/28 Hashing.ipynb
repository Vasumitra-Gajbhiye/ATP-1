## Performance Characteristics
The performance of sets and dictionaries in Python is a major reason for their widespread use. By leveraging hash tables, these data structures provide efficient storage and retrieval capabilities. Here we will go into the performance characteristics of sets and dictionaries, focusing on average-case and worst-case scenarios, as well as the strategies employed to maintain performance.

### Lookup
The average-case performance for lookups in sets and dictionaries is O(1). When you look up a key in a dictionary or check membership in a set, Python computes the hash value of the key and uses it to index directly into the hash table.

```python
my_dict = {'apple': 1, 'banana': 2, 'cherry': 3}
print(my_dict['banana'])
```

### Insertion and Deletion

Insertion and deletion operations also have an average-case time complexity of O(1). When inserting a new key-value pair into a dictionary or adding an element to a set, Python calculates the hash value and places the element in the corresponding bucket. Deletion follows a similar process by locating the element through its hash value and removing it.

```python
my_dict = {'apple': 1, 'banana': 2, 'cherry': 3}
my_dict['date'] = 4
del my_dict['banana']
print(my_dict)
```

# Find Maximum sum of the subarray (contigious locations) of size k

### Bruteforce approach

The brute force method checks every possible contiguous subarray of size k, computes its sum from scratch, and keeps track of the maximum sum found.

```python
def max_sum_brute_force(arr, k):
    n = len(arr)
    if n < k or k <= 0:
        return None  # Edge case: array is smaller than window size

    max_sum = float('-inf') #if we have all numbers in the array as negative

    # Loop through all valid starting indices
    for i in range(n - k + 1):
        current_sum = 0

        # Sum the next k elements
        for j in range(i, i + k):
            current_sum += arr[j]
        max_sum = max(max_sum, current_sum)

    return max_sum
```

```python
arr = [2, 1, 5, 1, 3, 2]
k = 3
print("Brute Force Max Sum:", max_sum_brute_force(arr, k))
```

```python
def max_sum_sliding_window(arr, k):
    n = len(arr)
    if n < k or k <= 0:
        return None  # Edge case: array is smaller than window size

    # Compute the sum of the first window of size k
    window_sum = sum(arr[:k])
    max_sum = window_sum

    # Slide the window from index k to the end of the array
    for i in range(k, n):
        # Add the incoming element and remove the outgoing element
        window_sum += arr[i] - arr[i - k]
        max_sum = max(max_sum, window_sum)

    return max_sum
```

```python
arr = [2, 1, 5, 1, 3, 2]
k = 3
print("Sliding Window Max Sum:", max_sum_sliding_window(arr, k))
```

```python
import random
import time

# Generate large test data
data = [random.randint(-100, 100) for _ in range(10000)]
k = 500

# Time Brute Force
start = time.perf_counter()
bf_ans = max_sum_brute_force(data, k)
bf_time = time.perf_counter() - start

# Time Sliding Window
start = time.perf_counter()
sw_ans = max_sum_sliding_window(data, k)
sw_time = time.perf_counter() - start

# Print results
print(f"Brute Force:    {bf_time:.4f}s (Answer: {bf_ans})")
print(f"Sliding Window: {sw_time:.4f}s (Answer: {sw_ans})")
```

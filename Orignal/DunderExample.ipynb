# Magic Methods = Dunder methods (double undersore methods).
- There are many Dunder methods which get automatically called when we access some of the built in operations.

- This allows developers to define or customise the behaviours of the objects.

````python
class Book:
    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages
````

````python
book1 = Book("Harry Potter", "J.K. Rowling", 500)
book2 = Book("Lord of the Rings", "J.R.R. Tolkien", 700)
````

````python
print(book1)
````

When we print book we get memory location of book1 object of class Book.

But what if when I `print(book1)` I want it to print details like this - `Harry Potter by J.K. Rowling`

````python
class Book:
    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages

    def __str__(self):
        return f"{self.title} by {self.author}"
````

````python
book1 = Book("Harry Potter", "J.K. Rowling", 500)
book2 = Book("Lord of the Rings", "J.R.R. Tolkien", 700)
````

````python
print(book1)
````

Creating Lists, Indexing & Negative Indexing

A **list** is an ordered, changeable collection of items, written with square brackets `[]`.

```python
fruits = ["apple", "banana", "cherry"]
```

Lists can hold any data type, even a mix of types, and indexing works just like strings — starting at `0`, with negative indices counting from the end.

````python
fruits = ["apple", "banana", "cherry", "date", "Fig"]

print(fruits)
print(fruits[0])     # first element
print(fruits[2])     # third element
print(fruits[-1])    # last element
print(fruits[-2])    # second-to-last element
print(len(fruits))   # number of elements
````

````python
mixed = ["Mark", 16, 5.4, True]   # a list can mix types
print(mixed)

nested = [1, 2, [3, 4, 5]]          # a list can even contain another list
print(nested)
print(nested[2])
print(nested[2][0])                 # access an item inside the nested list
````

## Traversal — Looping Over Lists

There are two common ways to loop over a list:

**By value** (most common):
```python
for item in my_list:
    print(item)
```

**By index** (when you need the position too):
```python
for i in range(len(my_list)):
    print(i, my_list[i])
```

````python
fruits = ["apple", "banana", "cherry"]

print("By value:")
for fruit in fruits:
    print(fruit)

print("\nBy index:")
for i in range(len(fruits)):
    print(i, fruits[i])
````

## Mutability — Add / Remove / Update Elements

Unlike strings, lists are **mutable** — you can change them after creation.

| Operation | Code | Effect |
|---|---|---|
| Update by index | `lst[i] = value` | replaces the element at index `i` |
| Add to many elements | `lst.extend(value)` | iterable addition |
| Add to end | `lst.append(value)` | adds one item to the end |
| Insert at position | `lst.insert(i, value)` | inserts `value` at index `i` |
| Remove by value | `lst.remove(value)` | removes the first matching item |
| Remove by index | `lst.pop(i)` | removes and returns item at index `i` (default: last item) |
| Remove all | `lst.clear()` | empties the list |
| Sort | `lst.sort()` | sorts the list in place |
| Reverse | `lst.reverse()` | reverses the list in place |
| Check membership | `value in lst` | returns `True`/`False` |


````python
numbers = [3, 1, 4, 1, 5, 9, 2]
print("Original:", numbers)

numbers[0] = 100          # update
print("After update:", numbers)

numbers.append(7)         # add to end
print("After append:", numbers)

numbers.insert(1, 50)     # insert at position 1
print("After insert:", numbers)

numbers.remove(1)         # removes the FIRST '1' it finds
print("After remove(1):", numbers)

popped = numbers.pop()    # removes and returns the last item
print("Popped value:", popped)
print("After pop:", numbers)

numbers.sort()
print("Sorted:", numbers)

numbers.reverse()
print("Reversed:", numbers)

print(100 in numbers)
````

### Question what is the difference between .insert and direct assignment
```lst.insert(i, value)``` and ```lst[i] = value```

```lst[i] = value``` (Assignment)

- Overwrites the existing element at index i.
- The list's length stays the same.
- If i is out of range, it raises an IndexError.

````python
lst = [10, 20, 30]
lst[1] = 99
print(lst)
````

```lst.insert(i, value)``` (Insertion)

- Inserts a new element at index i, pushing everything from that index onward one position to the right.
- The list's length increases by 1.
- If i is out of range (too large), it just inserts at the end — no error.

````python
lst = [10, 20, 30]
lst.insert(1, 99)
print(lst)
````

### In Summary
| Assignment | Insert |
|---|---|
| Effect Replaces element at i | Inserts new element at i |
| Length change	No change | Increases by 1 |
| Out-of-range index Raises IndexError | Inserts at end (no error) |

## Question what is the difference between .extend and .append methods

```lst.append()``` and ```lst.extend()```

````python
lst = [1, 2, 3]
lst.append([4, 5])
print(lst)

lst = [1, 2, 3]
lst.extend([4, 5])
print(lst)
````

### In Summary
| append | extend |
|---|---|
| Adds x as a single element to the end of the list, no matter what x is (even if it's a list itself) | Takes an iterable (list, tuple, string, etc.) and adds each of its elements individually to the end of the list |
| List length increases by exactly 1 | List length increases by the number of elements in the iterable |

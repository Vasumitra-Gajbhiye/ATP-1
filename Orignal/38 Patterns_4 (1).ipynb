## Total sales between two days

> You have daily sales for many days. Managers ask many questions of the form **"What were the total sales from day `l` to day `r`?"** (both days included). Answer each one.

### Step 1: Understand
`sales = [12, 30, 25, 8, 40, 35, 10, 22, 18, 5]` (days 0 to 9)
- Query (2, 5) → 25 + 8 + 40 + 35 = **108**
- Query (0, 9) → **205**
- Query (4, 4) → **40** (one day)

### Step 2: Decompose
(a) pick days `l` to `r`, (b) add them, (c) return the total.

### Step 3: Pseudocode
```

4. return total
```

### Analyse
Each query adds up to **n** days, so one query costs **O(n)**. With **q** queries, the total is **O(q × n)**.

**What if?:** *"If managers ask 10,000 questions, are we adding the same days again and again?"*
Yes. Ex: Day 5 is added in thousands of queries.

### Step 4 again: Avoid repeated work
Idea: **work out running totals once.** `prefix[i]` = total of the first `i` days.

| days added | prefix |
|---|---|
| day 0 | `prefix[0] = 12` |
| days 0-1 | `prefix[1] = 42` |
| ... | ... |

Then the total from day `l` to day `r` is:

> **prefix[r] − prefix[l-1]**   

### Predict, then run
Build the prefix list for `[3, 1, 4, 1, 5]` by hand. Then use it to find the total of days 1 to 3.

````python
def build_prefix(sales):
    prefix = [sales[0]]
    for i in range(1, len(sales)):
        prefix.append(prefix[-1] + sales[i])
    return prefix

def range_sum_fast(prefix, l, r):
    if l == 0:
        return prefix[r]
    return prefix[r] - prefix[l - 1]
````

````python
p = build_prefix([3, 1, 4, 1, 5])
print(p)
print(range_sum_fast(p, 1, 3))
````

### Compare brute force and prefix sums

````python
import random
import time
````

````python
prefix = build_prefix(sales)
for l, r in [(2, 5), (0, 9), (4, 4), (0, 0)]:
    print((l, r), range_sum_brute(sales, l, r), range_sum_fast(prefix, l, r))

# Many days, many queries
big = [random.randint(1, 100) for _ in range(100000)]
queries = [(random.randrange(50000), random.randrange(50000, 100000)) for _ in range(2000)]

start = time.perf_counter()
answers_brute = [range_sum_brute(big, l, r) for l, r in queries]
t_brute = time.perf_counter() - start

start = time.perf_counter()
big_prefix = build_prefix(big)
answers_fast = [range_sum_fast(big_prefix, l, r) for l, r in queries]
t_fast = time.perf_counter() - start

print("same answers:", answers_brute == answers_fast)
print(f"brute force : {t_brute:.4f} s")
print(f"prefix sums : {t_fast:.4f} s")
````

| Approach | Time for q queries | Extra memory |
|---|---|---|
| Brute force | O(q × n) | O(1) |
| Prefix sums | **O(n + q)** | O(n) |

**Discussion:** what if there is only **one** query? Then brute force is fine; building the prefix list costs more than it saves. Preparation only pays off when you ask **many** questions.

### The pattern map

| Pattern | Look for this signal | Idea | Typical cost |
|---|---|---|---|
| **Hashing** (dict/set) | "Have I seen this?", counting, matching pairs | Remember items for instant lookup | O(n) |
| **Prefix sums** | Many range-total questions | Pre-compute running totals | O(n + q) |
| **Two pointers** | Sorted data, a pair that meets a condition, partitioning | Two positions that move toward each other | O(n) |
| **Sliding window** | "Best/longest run of k consecutive items" | Add the new item, drop the old one | O(n) |
| **Binary search** | Sorted data, "find or decide fast" | Throw away half each step | O(log n) |
| **Memoization / DP** | The same sub-problem repeats | Store sub-answers | Varies |
| **Greedy** | A safe "best next choice" | Commit to the best local choice | Often O(n log n) |

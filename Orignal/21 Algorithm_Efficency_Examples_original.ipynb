## Activity - Check the Complexity of this code

### Calculate Order of Operations

```python
def int_to_string(number): # log?
    digits = '0123456789'
    if number == 0:
        return '0'
    result = ''
    while number > 0:
        result = digits[number % 10] + result
        number //= 10
    return result

print(int_to_string(12345))
```

```python
n = 1000
i, j, k = 0, 0, 0 #n log n

for i in range(n // 2, n + 1):
    j = 2
    while j <= n:
        k = k + n // 2
        j = j * 2

print("Final k =", k)
```

```python
def factorial(n): #n
    if n == 1:
        return 1
    else:
        return n * factorial(n - 1)

factorial(4)
```

```python
def fib(n):
    if n == 0 or n == 1:
        return n
    return fib(n - 1) + fib(n - 2)
```

```python
import time

for n in [1, 10, 20, 30, 40]:
    start = time.time()
    result = fib(n)
    elapsed = time.time() - start
    print(f"fib({n}) = {result}  |  took {elapsed:.4f} seconds")
```

## Activity - Check the Complexity of this code

### Calculate Order of Operations - Fibonacci example with memoization
This is the core idea behind **Dynamic Programming**: break a problem into overlapping sub-problems, solve each sub-problem only once, and reuse the stored answer everywhere it's needed again.

```python
def fibonacci(months, val_dict):
    if months in val_dict:
        return val_dict[months]
    else:
        val_dict[months] = fibonacci(months-1, val_dict) + fibonacci(months-2, val_dict)
        return val_dict[months]
```

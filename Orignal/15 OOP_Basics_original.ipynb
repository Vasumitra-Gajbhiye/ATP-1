So far we've stored related data using dictionaries, tuples aor any other datatypes, and behaviour using functions — but they were separate. **Object-Oriented Programming (OOP)** lets us bundle **data (attributes)** and **behaviour (methods)** together into one thing: an **object**.

## What is a class?
Class is the Blueprint

Ex: `User Class`
- "What it takes"  -->  These are the **Attributes** (the data variables like username and password).
- "What it checks for and what message it gives"  --> These are the **Methods**.

### Ex: Cookie Cutter and Cookies

- A **class** is like a **cookie cutter** — a template/blueprint that defines the *shape* every cookie will have.
- An **object** is an actual **cookie** made using that cutter — a specific instance with its own values.

You can make many cookies (objects) from one cutter (class), and each cookie can be decorated differently (have different attribute values), but they all share the same basic shape (structure).

Class:
 - Attributes: Data/Properties
 - Methods: Functions/Behavior

## Bubble sort - Make sort code Adaptable

````python
def bubble_sort(data):
  for i in range(len(data)-1):
    for j in range(len(data)-1-i):
      if data[j] > data[j+1]:
        data[j], data[j+1] = data[j+1], data[j]
  return data
````

````python
data = [2, 4, 5, 6, 7, 8, 9]
bubble_sort(data)
````

# Selection Sort

````python
def selection_sort(data):

    # Loop with index i
    for i in range(len(data) - 1):

        #set min value as i
        min = i

        for j in range(i+1,len(data)):
            if data[j] < data[min]:
                min = j

        #Since you have now got minimum value - swap and palce it at index i
        data[i],data[min] = data[min],data[i]

    return data
````

````python
arr = [23, 12, 34, 11, 100, 56, 78]
selection_sort(arr)
````

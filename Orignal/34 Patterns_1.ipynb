## Check for Duplicate transaction IDs

> A list of transaction IDs. Return `True` if **any ID appears more than once**, otherwise `False`.

### Step 1: Understand
- **Input:** a list of IDs.
- **Output:** `True` or `False`.
- Example 1: `[4021, 5307, 4488, 5307, 4021, 9100]` → **True** (5307 and 4021 repeat)
- Example 2: `[4021, 5307, 4488, 9100]` → **False**

### Step 2: Decompose
(a) take one ID, (b) check whether it appears again, (c) repeat for every ID.

### Step 3: Pseudocode
```
1. for each position i
2.     for each later position j
3.         if the IDs at i and j are equal, return True
4. return False
```

### Step 4: Approach
**Brute force:** compare every pair.

### Step 5: Code


````python
# def dup_brute(ids):
#     newSet = set()
#     for id in ids:
#         if id in newSet:
#             return True
#         newSet.add(id)
#     return False

def dup_brute(ids):
    for i in range(len(ids)):
        for j in range(i + 1, len(ids)):
            if ids[i] == ids[j]:
                return True
    return False
````

````python
print(dup_brute([4021, 5307, 4488, 5307, 4021, 9100]))
print(dup_brute([4021, 5307, 4488, 9100]))
````

### Step 6: Test and analyse
Edge cases: empty list, one item, exactly two equal items.

````python
tests = [([], False), ([7], False), ([5, 5], True),
         ([4021, 5307, 4488, 9100], False),
         ([4021, 5307, 4488, 5307, 4021, 9100], True)]

for ids, expected in tests:
    result = dup_brute(ids)
    print("OK  " if result == expected else "FAIL", ids, "->", result)
````

### Predict, then run
For a list of **1,000 different IDs** (no duplicates), how many comparisons does `dup_brute` make in total?

````python
def count_comparisons(n):
    count = 0
    for i in range(n):
        for j in range(i + 1, n):
            count += 1
    return count

for n in [10, 100, 1000]:
    print(f"n = {n:>4}  comparisons = {count_comparisons(n):>7,}   n*(n-1)/2 = {n*(n-1)//2:>7,}")
````

The count is **n(n-1)/2**, so the cost is **O(n²)**. If n grows 10 times, the work grows about **100 times**.

### Can we avoid repeated work?
**Question:** "What would make duplicates easy to spot?"


````python
def dup_check_1(ids):
    s = sorted(ids)
    for i in range(len(s) - 1):
        if s[i] == s[i + 1]:
            return True
    return False
````

````python
print(dup_check_1([4021, 5307, 4488, 5307, 4021, 9100]))
print(dup_check_1([4021, 5307, 4488, 9100]))
````

````python
def dup_check_2(ids):
    seen = set()
    for id_ in ids:
        if id_ in seen:
            return True
        seen.add(id_)
    return False
````

````python
print(dup_check_2([4021, 5307, 4488, 5307, 4021, 9100]))
print(dup_check_2([4021, 5307, 4488, 9100]))
````

### Compare all methods on the same tests

````python
import random, time

for f in (dup_brute, dup_check_1, dup_check_2):
    ok = all(f(ids) == expected for ids, expected in tests)
    print(f"{f.__name__:<10} passes all tests: {ok}")

ids = random.sample(range(10**7), 4000)
print()
for f in (dup_brute, dup_check_1, dup_check_2):
    start = time.perf_counter()
    f(ids)
    print(f"{f.__name__:<10} {time.perf_counter() - start:.5f} seconds")
````

### Build it
Write `find_duplicates(ids)` that returns a **sorted list of every ID that appears more than once** (each listed once).

````python
def find_duplicates(ids):
    seen = set()
    dups = set()
    for id in ids:
        if id in seen:
            dups.add(id)
        else:
            seen.add(id)
    return sorted(dups)

print(find_duplicates([4021, 5307, 4488, 5307, 4021, 9100]))
print(find_duplicates([]))
print(find_duplicates([1, 1, 1]))
````

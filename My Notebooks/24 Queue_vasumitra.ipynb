# Queues

## The queue idea

### What is a queue?
A queue follows **FIFO: First In, First Out.** The first item to arrive is the first one served, like people in a line.

| Operation | What it does |
|---|---|
| `enqueue(item)` | Join the **back** of the line |
| `dequeue()` | Remove and return the item at the **front** |
| `peek()` | Look at the front item **without** removing it |
| `is_empty()` | Is the line empty? |

### Lets build a Queue using linked list

For a queue, we remove at the **Begining** and add at the **End**.

| Operation | linked list |
|---|---|
| Remove at the **head** (dequeue)| O(1) |
| Add at the **tail** (enqueue)| O(n) as you have traverse till end|

````python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
````

````python
class Queue:
    def __init__(self):
        self.head = None
        self._size = 0

    def is_empty(self):
        return self.head is None

    def enqueue(self, item):
        node = Node(item)
        if self.head is None:
            self.head = node
        else:
            current = self.head
            while current.next is not None:   # walk until the last node
                current = current.next
            current.next = node               # attach after it
        self._size += 1

    def dequeue(self):
        if self.is_empty():
            return None
        item = self.head.data
        self.head = self.head.next
        self._size -= 1
        return item

    def peek(self):
        if self.is_empty():
            return None
        return self.head.data

    def size(self):
        return self._size

    def __str__(self):
        items = []
        current = self.head
        while current is not None:
            items.append(str(current.data))
            current = current.next
        return "FRONT [" + ", ".join(items) + "] BACK" if items else "(empty queue)"
````

Now if the `Queue` keeps **two** pointers: `head` (front) and `tail` (back). rewrite `enqueue` and `dequeue`.

| Operation | linked list |
|---|---|
| Remove at the **head** | O(1) |
| Add at the **head** | O(1) |
| Add at the **tail**  with tail pointer | O(1) |


````python
class Queue:
    def __init__(self):
        self.head = None        # FRONT of the line
        self.tail = None        # BACK of the line
        self._size = 0

    def is_empty(self):
        return self.head is None

    def enqueue(self, item):
        node = Node(item)
        if self.tail is None:       # queue is empty: node is both front and back
            self.head = node
            self.tail = node
        else:
            self.tail.next = node   # attach after the current last node
            self.tail = node        # tail now points at the new last node
        self._size += 1

    def dequeue(self):
        if self.is_empty():
            return None
        item = self.head.data
        self.head = self.head.next  # front moves to the next node
        if self.head is None:       # the queue just became empty...
            self.tail = None        # ...so tail must be reset too
        self._size -= 1
        return item

    def peek(self):
        if self.is_empty():
            return None
        return self.head.data

    def size(self):
        return self._size

    def __str__(self):
        items = []
        current = self.head
        while current is not None:
            items.append(str(current.data))
            current = current.next
        return "FRONT [" + ", ".join(items) + "] BACK" if items else "(empty queue)"
````

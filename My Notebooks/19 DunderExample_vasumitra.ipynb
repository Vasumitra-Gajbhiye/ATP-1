# Magic Methods = Dunder methods (double undersore methods).
- There are many Dunder methods which get automatically called when we access some of the built in operations.

- This allows developers to define or customise the behaviours of the objects.

````python
class Book:
    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages
````

````python
book1 = Book("Harry Potter", "J.K. Rowling", 500)
book2 = Book("Lord of the Rings", "J.R.R. Tolkien", 700)
````

````python
print(book1)
````

When we print book we get memory location of book1 object of class Book.

But what if when I `print(book1)` I want it to print details like this - `Harry Potter by J.K. Rowling`

````python
class Book:
    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages

    def __str__(self):
        return f"{self.title} by {self.author}"
````

````python
book1 = Book("Harry Potter", "J.K. Rowling", 500)
book2 = Book("Lord of the Rings", "J.R.R. Tolkien", 700)
````

````python
print(book1)
````

# Activity

Update the class above such that when we check
`book1 == book2` it should give `True` if name and author are same.

Note: magic method of == is `__eq__()`

````python
class Book:
    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages

    def __str__(self):
        return f"{self.title} by {self.author}"

    def __eq__(self, other):
        if not isinstance(other, Book):
            return NotImplemented
        return self.title == other.title and self.author == other.author
````

````python
book1 = Book("Harry Potter", "J.K. Rowling", 500)
book2 = Book("Lord of the Rings", "J.R.R. Tolkien", 700)
book3 = Book("Harry Potter", "J.K. Rowling", 320)  # same title & author, different pages

print(book1 == book2)  # False
print(book1 == book3)  # True
````

# Activity - Build

Overload "+" operator such that we can use it on 2 equal length sets.

```
set1 = {1, 2, 3}
set2 = {2, 4, 6}

set1 + set2
```
Note: magic method is `__add__()`

````python
# We can't change how + works on Python's built-in set,
# so we create our own class that inherits from set and overrides __add__.
class MySet(set):
    def __add__(self, other):
        if len(self) != len(other):
            raise ValueError("Both sets must be of equal length")
        # sets are unordered, so sort both and add element-wise
        return MySet(a + b for a, b in zip(sorted(self), sorted(other)))
````

````python
set1 = MySet({1, 2, 3})
set2 = MySet({2, 4, 6})

set1 + set2  # {3, 6, 9}
````

````python
# Unequal lengths raise an error
try:
    MySet({1, 2}) + MySet({1, 2, 3})
except ValueError as e:
    print(e)
````

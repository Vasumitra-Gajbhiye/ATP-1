# Object Oriented (OO) programming

A Programming technique where code is organized as a collection of objects.

An object consists of:
- Data. Attributes
- Behavior. Methods which are functions applied to the object.

# Class and Object

Everything in Python is an object, with its properties and methods. A number, string, list, dictionary, tuple, set etc. used in a program is an object of a corresponding built-in class.

# What is a Class?
A class is like an object constructor, or a "blueprint" for creating objects. We instantiate a class to create an object. The class defines attributes and the behavior of the object, while the object, on the other hand, represents the class.

## Syntax to create class
```
class ClassName:
  code goes here
```
Class name should be **CamelCase.**

## Creating an Object
We can create an object by calling the class.
```
class Animal:
  pass
```
```
cat = Animal()
```

````python
john = Atm()
try:
    john.menu()
except TypeError as e:
    print("TypeError:", e)
````

**Why does `john.menu()` fail?**

`TypeError: Atm.menu() takes 0 positional arguments but 1 was given`

`john.menu()` is really `Atm.menu(john)`: Python passes the object itself as the first argument. Since `menu()` was defined without a parameter, there is nowhere to put `john`. **Fix:** every method must take `self` as its first parameter → `def menu(self):`

A class has a method, now this methods cannot talk to another method in same class. Only object has access to methods of a class. Hence if say menu method want to access check balance method it should refer it as self.check_balance()

````python
class Atm:
    # we always define data within a sepcial method called __init__
    # __init__ is a constructor
    # Special method inside a class.
    # Initializes instance members.
    # Executes automatically on object creation.
    def __init__(self):
        self.pin = ""
        self.balance = 0

        self.menu()

    def menu(self):
        # keep showing the menu until the user chooses to exit
        while True:
            user_input = input("""
          Hi how can I help you?
          1. Press 1 to create pin
          2. Press 2 to deposit
          3. Press 3 to withdraw
          4. Press 4 to check balance
          5. Anything else to exit""")

            if user_input == "1":
                self.create_pin()
            elif user_input == "2":
                self.deposit()
            elif user_input == "3":
                self.withdraw()
            elif user_input == "4":
                self.check_balance()
            else:
                print("Thank you for transacting with us.")
                break

    def create_pin(self):
        self.pin = input("Enter your pin: ")
        print("Pin set successfully")

    def deposit(self):
        temp = input("Enter your pin: ")
        if temp == self.pin:
            amount = int(input("Enter the amount: "))
            self.balance = self.balance + amount
            print("Deposit successful")
        else:
            print("Invalid pin")

    def withdraw(self):
        temp = input("Enter your pin: ")
        if temp == self.pin:
            amount = int(input("Enter the amount: "))
            if amount <= self.balance:
                self.balance = self.balance - amount
                print("Operation successful")
            else:
                print("insufficient funds")
        else:
            print("invalid pin")

    def check_balance(self):
        temp = input("Enter your pin: ")
        if temp == self.pin:
            print(self.balance)
        else:
            print("invalid pin")
````

Creating an object runs `__init__`, which opens the menu. Try: `1` → set PIN, `2` → deposit, `3` → withdraw, `4` → check balance, `5` → exit.

````python
sbi = Atm()
````

## Encapsulation: private members

Putting `__` in front of an attribute or method name makes it private. Python renames it internally (`__balance` → `_Atm__balance`, called *name mangling*), so code outside the class can't reach it by its plain name.

Inside the class, every method must use the same private name (`self.__balance`, not `self.balance`).

````python
class Atm:
    # we always define data within a sepcial method called __init__
    # __init__ is a constructor
    # Special method inside a class.
    # Initializes instance members.
    # Executes automatically on object creation.
    def __init__(self):
        # __ makes these private: they can't be reached from outside the class
        self.__pin = ""
        self.__balance = 0

        self.__menu()

    # Getters and setters: the controlled "doors" to the private data
    def get_balance(self):
        return self.__balance

    def set_balance(self, amount):
        # Setter with validation: the balance must be a non-negative number
        if not isinstance(amount, (int, float)) or amount < 0:
            print("Balance must be a non-negative number")
            return False
        self.__balance = amount
        return True

    def __menu(self):
        # keep showing the menu until the user chooses to exit
        while True:
            user_input = input("""
          Hi how can I help you?
          1. Press 1 to create pin
          2. Press 2 to deposit
          3. Press 3 to withdraw
          4. Press 4 to check balance
          5. Anything else to exit""")

            if user_input == "1":
                self.__create_pin()
            elif user_input == "2":
                self.__deposit()
            elif user_input == "3":
                self.__withdraw()
            elif user_input == "4":
                self.__check_balance()
            else:
                print("Thank you for transacting with us.")
                break

    def __create_pin(self):
        self.__pin = input("Enter your pin: ")
        print("Pin set successfully")

    def __deposit(self):
        temp = input("Enter your pin: ")
        if temp == self.__pin:
            amount = int(input("Enter the amount: "))
            self.__balance = self.__balance + amount
            print("Deposit successful")
        else:
            print("Invalid pin")

    def __withdraw(self):
        temp = input("Enter your pin: ")
        if temp == self.__pin:
            amount = int(input("Enter the amount: "))
            if amount <= self.__balance:
                self.__balance = self.__balance - amount
                print("Operation successful")
            else:
                print("insufficient funds")
        else:
            print("invalid pin")

    def __check_balance(self):
        temp = input("Enter your pin: ")
        if temp == self.__pin:
            print(self.__balance)
        else:
            print("invalid pin")
````

````python
sbi = Atm()
````

Accessing private members from outside raises `AttributeError`. (`try`/`except` catches the error so the rest of the notebook still runs.)

````python
try:
    sbi.__balance
except AttributeError as e:
    print("AttributeError:", e)
````

````python
try:
    sbi.__deposit()
except AttributeError as e:
    print("AttributeError:", e)
````

Name mangling is not real security: the renamed attribute is still reachable. Python relies on convention, so never touch `__` members from outside the class.

````python
sbi._Atm__balance
````

## The right way: getters and setters

The setter checks the value first, so `"hello"` or a negative balance is rejected.

````python
print(sbi.set_balance(1000))
print(sbi.get_balance())
print(sbi.set_balance("hello"))
print(sbi.set_balance(-50))
print(sbi.get_balance())
````

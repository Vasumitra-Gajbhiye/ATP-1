# Algorithmic Complexity

## Finding duplicates in a list.

### Solution 1

```python
def has_duplicates(items):

    for i in range(len(items)):

        for j in range(i + 1, len(items)):
            if items[i] == items[j]:
                return True

    return False
```

```python
list1 = [1, 7, 90, 100, 66, 74, 55, 44, 1, 90]

has_duplicates(list1)
```

### Solution 2

```python
def has_duplicates(items):
    seen = set()

    for item in items:

        if item in seen:
            return True

        seen.add(item)

    return False
```

```python
list1 = [1, 7, 90, 100, 66, 74, 55, 44, 1, 90]

has_duplicates(list1)
```

### Solution 3

```python
def has_duplicates(items):
    # a set keeps only unique values, so if the size shrinks there were duplicates
    return len(set(items)) != len(items)
```

```python
list1 = [1, 7, 90, 100, 66, 74, 55, 44, 1, 90]

has_duplicates(list1)
```

```python
# Complexity comparison
# Solution 1 : nested loops            -> O(n^2) time, O(1) extra space
# Solution 2 : set with early exit     -> O(n)   time, O(n) extra space (stops at first duplicate)
# Solution 3 : len(set) vs len(list)   -> O(n)   time, O(n) extra space (always builds the full set)
```

# Measuring Time

Let's say I want to print 1 to 100 numbers. How will be the code along with code to measure time.

```python
import time

start = time.time()

for i in range(1, 101):
    print(i)

end = time.time()

print("Time taken:", end - start, "seconds")
```

```python
# perf_counter() is a high-resolution clock, better suited for measuring short durations
start = time.perf_counter()

for i in range(1, 101):
    print(i)

end = time.perf_counter()

print("Time taken:", end - start, "seconds")
```

```python
# Comparing the three duplicate-finding solutions on a large list with no duplicates (worst case)

def has_duplicates_1(items):
    for i in range(len(items)):
        for j in range(i + 1, len(items)):
            if items[i] == items[j]:
                return True
    return False


def has_duplicates_2(items):
    seen = set()
    for item in items:
        if item in seen:
            return True
        seen.add(item)
    return False


def has_duplicates_3(items):
    return len(set(items)) != len(items)


big_list = list(range(5000))

for func in [has_duplicates_1, has_duplicates_2, has_duplicates_3]:
    start = time.perf_counter()
    result = func(big_list)
    end = time.perf_counter()
    print(func.__name__, result, f"{end - start:.6f} seconds")
```

```python
# timeit runs the code many times and gives a more reliable measurement
import timeit

print("Solution 1:", timeit.timeit(lambda: has_duplicates_1(big_list), number=1))
print("Solution 2:", timeit.timeit(lambda: has_duplicates_2(big_list), number=100) / 100)
print("Solution 3:", timeit.timeit(lambda: has_duplicates_3(big_list), number=100) / 100)
```

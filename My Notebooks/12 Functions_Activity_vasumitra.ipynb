# Activity: Warehouse Order Processing - Using Functions

You've already built the Warehouse Order Processing System using plain variables and conditionals. In this activity, you'll **rewrite it using functions**.

**Rules for this activity:**
- Every function you write should **`print()`** its result directly.
- Each function should do **one job only** — don't combine multiple checks into a single function.
- Crete a final function, `process_order()`, which will **call** all your other functions in sequence to produce the full order summary. This is called **function composition**.

**Business rules (same as before):**
- Shipping cost = ₹50 base charge + ₹8 per km of delivery distance.
- An order is **ACCEPTED** only if: order value > ₹500, at least 1 item, and distance ≤ 100 km. Otherwise it's **REJECTED**.
- Shipping is **FREE** if order value ≥ ₹2000 OR the customer is a premium member.
- Processing is **PRIORITY** only if the order is marked priority AND its value is ≥ ₹1500. Otherwise it's **STANDARD**.
- Handling is **SPECIAL** if the order contains a fragile item OR has 10+ items. Otherwise it's **NORMAL**.

## Test Data

Use this list of orders to test your functions once they're written. Each order is a tuple:

`(customer_id, order_value, num_items, distance_km, is_premium, is_priority, is_fragile)`

This test set is designed to cover multiple cases.

````python
test_orders = [
    ("C101", 2450, 6, 18, True, False, True),
    ("C102", 300, 2, 50, False, False, False),
    ("C103", 1600, 12, 30, False, True, False),
    ("C104", 5000, 1, 120, True, True, False),
    ("C105", 800, 0, 10, False, False, False),
]
````

## Function: `display_order_info`

**Parameters:** `customer_id, order_value, num_items, distance_km`

**What it should do:** Print the order's basic details, one field per line, in this exact format:
```
Customer          : C101
Order value       : Rs. 2450
Items             : 6
Distance          : 18 km
```

````python
def display_order_info(customer_id, order_value, num_items, distance_km):
    print(f"Customer          : {customer_id}")
    print(f"Order value       : Rs. {order_value}")
    print(f"Items             : {num_items}")
    print(f"Distance          : {distance_km} km")
````

## Function: `check_order_status`

**Parameters:** `order_value, num_items, distance_km`

**What it should do:** Apply the acceptance rule and print either:
```
Status            : ACCEPTED
```
or
```
Status            : REJECTED
```

````python
def check_order_status(order_value, num_items, distance_km):
    if order_value > 500 and num_items >= 1 and distance_km <= 100:
        print("Status            : ACCEPTED")
    else:
        print("Status            : REJECTED")
````

## Function: `check_shipping_status`

**Parameters:** `order_value, is_premium, distance_km`

**What it should do:**
- Calculate the shipping cost (₹50 base + ₹8/km).
- If the order qualifies for free shipping, print:
```
Shipping          : FREE
```
- Otherwise print the calculated cost, e.g.:
```
Shipping          : Rs. 450
```

````python
def check_shipping_status(order_value, is_premium, distance_km):
    shipping_cost = 50 + 8 * distance_km
    if order_value >= 2000 or is_premium:
        print("Shipping          : FREE")
    else:
        print(f"Shipping          : Rs. {shipping_cost}")
````

## Function: `check_priority_status`

**Parameters:** `is_priority, order_value`

**What it should do:** Print either:
```
Processing        : PRIORITY
```
or
```
Processing        : STANDARD
```

````python
def check_priority_status(is_priority, order_value):
    if is_priority and order_value >= 1500:
        print("Processing        : PRIORITY")
    else:
        print("Processing        : STANDARD")
````

## Function: `check_handling_status`

**Parameters:** `is_fragile, num_items`

**What it should do:** Print either:
```
Handling          : SPECIAL
```
or
```
Handling          : NORMAL
```

````python
def check_handling_status(is_fragile, num_items):
    if is_fragile or num_items >= 10:
        print("Handling          : SPECIAL")
    else:
        print("Handling          : NORMAL")
````

## Function: `process_order` (Composition)

**Parameters:** `customer_id, order_value, num_items, distance_km, is_premium, is_priority, is_fragile`

**What it should do:** This function doesn't calculate anything itself — it just **calls your other 5 functions in the right order**, with a separator line and header before and after, to build the full summary:
```
===================================
   ORDER PROCESSING RESULT
===================================
Customer          : C104
Order value       : Rs. 2450
Items             : 6
Distance          : 18 km

Status            : ACCEPTED
Shipping          : FREE
Processing        : STANDARD
Handling          : SPECIAL
===================================
```

````python
def process_order(customer_id, order_value, num_items, distance_km, is_premium, is_priority, is_fragile):
    print("===================================")
    print("   ORDER PROCESSING RESULT")
    print("===================================")
    display_order_info(customer_id, order_value, num_items, distance_km)
    print()
    check_order_status(order_value, num_items, distance_km)
    check_shipping_status(order_value, is_premium, distance_km)
    check_priority_status(is_priority, order_value)
    check_handling_status(is_fragile, num_items)
    print("===================================")
````

## Run your functions against the test data

Once all 6 functions are written, run the cell below. It loops through `test_orders` and calls `process_order()` once for each order, using `*order` to unpack each tuple directly into the function's parameters.

````python
for order in test_orders:
    process_order(*order)
````

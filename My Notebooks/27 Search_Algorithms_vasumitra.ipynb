### Brute force

> **Brute force** means trying **every possibility** until one works. No cleverness, just effort.

It is the first idea most people have, and there is nothing wrong with that. The question is *how much effort* it costs as the input grows.

### Find a number in a list - O(n)

```python
def linear_search(items, target):
    steps = 0
    for i, x in enumerate(items):
        steps += 1
        if x == target:
            return i, steps
    return -1, steps
```

```python
nums = [42, 7, 19, 85, 3, 66, 28, 51, 90, 14]

for target in (42, 51, 14, 99):
    index, steps = linear_search(nums, target)
    print(f"search {target:>2}: index {index:>2}, looked at {steps} item(s)")
```

## Excercise: Does the list contain a duplicate?
Brute force approach: compare **every pair**.

```python
def has_duplicate_brute(items):
    # Compare every pair (i, j) with i < j. Stop as soon as a match is found.
    comparisons = 0
    n = len(items)
    for i in range(n):
        for j in range(i + 1, n):
            comparisons += 1
            if items[i] == items[j]:
                return True, comparisons
    return False, comparisons
```

```python
output = has_duplicate_brute([5, 3, 3, 8, 10, 3, 5, 8])
# print has duplicates or not and how many comparisions done?
found, comparisons = output
print(f"has duplicates: {found}, comparisons done: {comparisons}")

# Worst case (no duplicates): every pair is checked -> n*(n-1)/2 comparisons, O(n^2)
print(has_duplicate_brute([1, 2, 3, 4, 5, 6, 7, 8]))
```

# Binary search (Only on sorted List)

### Find a number in a list - O(log n)

#### The algorithm


*   Keep two markers, low and high, around the part that could still hold the target.
*   Look at the middle item.
    - If it equals the target, you are done.
    - If it is smaller than the target, the target is to the right: low = mid + 1.
    - If it is bigger, the target is to the left: high = mid - 1.
    - If low passes high, nothing is left to search: not found.

```python
def binary_search(numerlist, low, high, item):
    step=0
    if low > high:                      
        return -1
    mid = (low + high) // 2
    if numerlist[mid] == item:
        print(f"Step {step}")
        return mid
    elif numerlist[mid] < item:         # to right
        step+=1
        return binary_search(numerlist, mid + 1, high, item)
    else:                               # to left
        step +=1
        return binary_search(numerlist, low, mid - 1, item)
```

```python
numerlist = [12,24,35,46,57,68,80,99,100]

print(binary_search(numerlist, 0, len(numerlist) - 1, 99))
```


## Exercise: Find the First True in a Sorted Boolean Array

**Problem Statement:**
An array of boolean values is divided into two sections: The left section consists of all false, and the right section consists of all true. Find the first true element of the right section and return its index. If there is no true element in the array, return -1.
- Input: arr = [False, False, True, True, True]
- Output: 2
- Explanation: The first True appears at index 2.

```python
def find_boundary(datalist):
    low, high = 0, len(datalist) - 1
    boundary = -1
    while low <= high:
        mid = (low + high) // 2
        if datalist[mid]:
            boundary = mid
            high = mid - 1
        else:
            low = mid + 1
    return boundary
```

```python
datalist = [False, False, False, False, True, True, True]
result = find_boundary(datalist)
print(result)
```

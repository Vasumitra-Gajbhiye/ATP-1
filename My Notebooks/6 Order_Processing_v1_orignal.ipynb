# Warehouse Order Processing System

This notebook collects order details from the user, applies the warehouse's business rules, calculates shipping cost, and displays a summary of the order.

## Step 1: Collect order information from the user

```python
customer_id = input("Enter Customer ID: ").strip()
order_value = float(input("Enter Order value (₹): ").strip())
num_items  = int(input("Enter Number of items: ").strip())
distance_km = float(input("Enter Delivery distance (in km): ").strip())

# --- Premium member (Yes/No) ---
premium_input = input("Is the customer a premium member? (Yes/No): ").strip().lower()
is_premium = (premium_input == "yes")

# --- Priority order (Yes/No) ---
priority_input = input("Does the order require priority processing? (Yes/No): ").strip().lower()
is_priority = (priority_input == "yes")

# --- Fragile item (Yes/No) ---
fragile_input = input("Does the order contain a fragile item? (Yes/No): ").strip().lower()
is_fragile = (fragile_input == "yes")
```

## Step 2: Calculate shipping cost

Rule: ₹50 base charge + ₹8 per km of delivery distance

```python
base_charge = 50
per_km_charge = 8
shipping_cost = base_charge + (per_km_charge * distance_km)
```

## Step 3: Apply business rules using Boolean logic

```python
# Rule 1: Order can be processed only if value > 500,
# at least 1 item, and distance <= 100 km.
order_accepted = (order_value > 500) and (num_items >= 1) and (distance_km <= 100)

# Rule 2: Free shipping if order value >= 2000 OR premium member.
free_shipping = (order_value >= 2000) or (is_premium)

# Rule 3: Priority processing only if marked priority AND value >= 1500.
priority_processing = (is_priority) and (order_value >= 1500)

# Rule 4: Special handling if fragile item OR 10+ items.
special_handling = (is_fragile) or (num_items >= 10)
```

## Step 4: Convert boolean results into display labels

```python
if order_accepted:
    status_label = "ACCEPTED"
else:
    status_label = "REJECTED"

if not order_accepted:
    # If the order itself is rejected, shipping/processing/
    # handling don't really apply, but we still show
    # what would have been calculated for transparency.
    shipping_label = "NOT APPLICABLE"
    processing_label = "NOT APPLICABLE"
    handling_label = "NOT APPLICABLE"
else:
    if free_shipping:
        shipping_label = "FREE"
    else:
        shipping_label = "₹" + str(round(shipping_cost, 2))

    if priority_processing:
        processing_label = "PRIORITY"
    else:
        processing_label = "STANDARD"

    if special_handling:
        handling_label = "SPECIAL"
    else:
        handling_label = "NORMAL"
```

## Step 5: Display the order summary

A simple pattern generation loop is used to draw the separator lines instead of hardcoding the `=` characters.

```python
separator = ""
line_length = 35
counter = 0

# For loop used here to build the separator line
for counter in range(line_length):
    separator = separator + "="

print()
print(separator)
print("       ORDER PROCESSING RESULT")
print(separator)
print()

# Show order value as a whole number if it has no
# decimal part (e.g. 2450 instead of 2450.0)
if order_value == int(order_value):
    order_value_display = str(int(order_value))
else:
    order_value_display = str(order_value)

# Same idea for distance
if distance_km == int(distance_km):
    distance_display = str(int(distance_km))
else:
    distance_display = str(distance_km)

print("Customer          :", customer_id)
print("Order value       : ₹" + order_value_display)
print("Items             :", num_items)
print("Distance          :", distance_display, "km")
print()
print("Status            :", status_label)
print("Shipping          :", shipping_label)
print("Processing        :", processing_label)
print("Handling          :", handling_label)
print()
print(separator)
```

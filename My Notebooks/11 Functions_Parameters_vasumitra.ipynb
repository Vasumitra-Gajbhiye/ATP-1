## Why Functions? Defining and Calling

Imagine you had to write the same 5 lines of code every time you wanted to greet a user — 10 times in your program. That's repetitive and error-prone.

A **function** is a named, reusable block of code that performs a specific task. You **define** it once, and **call** it as many times as you like.

```python
def function_name(parameters):
    """ doc string """
    statements
```
- def           ---> Function start.
- function_name ---> Function identifier.
- parameters    ---> Input values.
- Colon (`:`)   ---> Ends header.
- doc string    ---> documentation.
- statements    ---> Function body.
- return        ---> Output value (optional).

```
function_name(values)
```


````python
def greet():
    """
    This function prints greeting message
    Input: None
    Output: None
    """
    print("Hello!")
    print("Welcome to the Python course.")

# Calling the function
greet()
greet()
greet()
````

````python
print(greet.__doc__)
````

````python
print.__doc__
````

### Why functions matter

- **Avoid repetition**
- **Organize code** into logical, named chunks
- **Easier to test and debug** — fix a bug in one place, not everywhere
- **Reusable** — write once, use anywhere in your program


````python
# Without a function - repetitive
print("=== Report ===")
print("Sales: 1000")
print("==============")

print("=== Report ===")
print("Expenses: 500")
print("==============")
````

````python
# With a function - cleaner
def print_report(label, value):
    print("=== Report ===")
    print(label + ":", value)
    print("==============")

print_report("Sales", 1000)
print_report("Expenses", 500)
````

## Parameters — Positional & Default Values

Functions become far more useful when they can accept **input values** — called **parameters**.

## Parameters Vs Arguments

**Parameters:**
- variables in () during function definition.
- Defined in function declaration.
```
def func(param1, param2):
    # Body
```

**Arguments:**
- Values passed at function call.
- Inputs during function invocation.
```
func(arg1, arg2)
```

````python
def greet(name):
    print("Hello,", name + "!")

greet("Aditi") # "Aditi" is the ARGUMENT passed in for the "name" PARAMETER
greet("Rohan")
greet("Meera")
````

Types of Argument in python:
1. Positional Argument
2. Default Argument
3. Keyword Argument
4. Arbitrary Argument (*args)

### Multiple parameters

Arguments are matched to parameters **in order** — this is called "positional" matching.


````python
def describe_student(name, age, grade):
    print(name, "is", age, "years old and is in grade", grade)

describe_student("Aditi", 16, "11th")
describe_student("Rohan", 15, "10th")

# ORDER MATTERS with positional arguments!
describe_student("10th", "Kabir", 14)   # this will look wrong
````

### Default Argument

You can give a parameter a **default value**, which is used if the caller doesn't provide one.

````python
def greet(name):
    print("Hello,", name + "!")

greet("Meera")
greet()
````

````python
def greet(name = "Stranger"):
    print("Hello,", name + "!")

greet("Meera")
greet()
````

````python
def greet(name, greeting="Hello"):
    print(greeting + ",", name + "!")

greet("Aditi")                    # uses the default greeting
greet("Rohan", "Good morning")    # overrides the default
````

Parameters with default values must come **after** parameters without defaults.

```python
def bad_example(name, greeting="Hello", age):   # ERROR - age has no default but comes after one that does
    print(name)
```


````python
def bad_example(name, greeting="Hello", age):
  print(name)
````

### Keyword Argument
You can explicitly pass keyword while calling the function which will oevrride the positional check.

````python
def greet(name, greeting="Hello"):
    print(greeting + ",", name + "!")

greet("Good evening", "John")
greet(greeting="Good evening", name="John") #positional override
````

### Arbitrary Argument
Arbitrary arguments in Python allow a function to accept an unknown or varying number of inputs during a function call. This is incredibly useful when you don't know ahead of time how many values a user might pass into your function.

- **Arbitrary Positional Arguments** (*args): lets your function accept any number of positional arguments. Python packs all the extra positional inputs into a tuple. Ex: print()

- **Arbitrary Keyword Arguments** (**kwargs): lets your function accept any number of keyworded (named) arguments. Python packs all the extra positional inputs into a standard dictionary.

````python
def product(*numbers):
  product_value = 1
  for i in numbers:
      product_value *= i
  print(product_value)

product(10)
product(2, 3, 4)
product(1, 2, 3, 4, 5, 9)
product()
````

````python
def display_profile(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

# You can pass as many named arguments as you want
display_profile(name="Alice", age=28, role="Developer")
display_profile(name="Jane", age=18)
display_profile(role="Intern", age=25)
display_profile(name="Mark")
display_profile()
````

### The Correct Argument Order

When defining a function, you must list your arguments in a specific order: positional arguments first, followed by *args, and then **kwargs last.

````python
def master_function(normal_arg, *args, **kwargs):
    print(f"Normal: {normal_arg}")
    print(f"args: {args}")
    print(f"kwargs: {kwargs}")

master_function("Hello", 1, 2, 3, status="Active", role="Admin")
````

The exact sequence you must follow in your function definition is:
1.   Positional arguments (Mandatory parameters without default values)
2.   Positional arguments with default values (Optional parameters)
3.   *args (Variable-length positional arguments)
4.   Keyword-only arguments (Parameters that must be passed by name, with or without defaults)
5.   **kwargs (Variable-length keyword arguments)

````python
def complete_sequence(
    pos_arg,                   # 1. Positional
    default_pos="Guest",       # 2. Positional with default
    *args,                     # 3. *args
    keyword_only,              # 4. Keyword-only (No default)
    default_keyword="Active",  # 4. Keyword-only (With default)
    **kwargs                   # 5. **kwargs
):
    print(f"1. Positional: {pos_arg}")
    print(f"2. Default Positional: {default_pos}")
    print(f"3. *args: {args}")
    print(f"4. Keyword-Only: {keyword_only}")
    print(f"5. Default Keyword: {default_keyword}")
    print(f"6. **kwargs: {kwargs}")

# --- Calling the function ---
complete_sequence(
    "Alice",              # Goes to pos_arg
    "Admin",              # Overrides default_pos
    100, 200, 300,        # taken into *args
    keyword_only="Yes",   # keyword argument (must use name)
    city="London",        # taken into **kwargs
    status="Online"       # taken into **kwargs
)
````

### Activity — Build it

Write a function called `make_username` that takes two parameters: `first_name` and a `separator` parameter that **defaults to** `"_"`.

### Activity — Problem 3: Temperature converter

Write a function `celsius_to_fahrenheit()` that prints the Fahrenheit equivalent, using the formula:

```
F = (C * 9/5) + 32
```
#### Test it with these values
- celsius = 0
- celsius = 100

````python
def celsius_to_fahrenheit(celsius):
    if celsius < -273.15:
        print("Error: Temperature below absolute zero")
    else:
        fahrenheit = (celsius * 9/5) + 32
        print(fahrenheit)

celsius_to_fahrenheit(0)     # 32.0
celsius_to_fahrenheit(100)   # 212.0
````

### Activity — Problem 4: Word counter with a default

Write a function `count_words()` that counts and prints how many words in the sentence have **at least** `min_length` characters. `min_length` should default to `1` (counting all words).

Example:
- Sentence = "If we have data, let us look at data. If all we have are opinions, let us go with mine"
- min_length = 2
- output = ["If", "we", "have", "data", "let", "us"]

````python
from this import d


def count_words(sentence, min_length=1):
    words = []
    for word in sentence.split():
        word = word.strip(".,!?;:")   # remove punctuation so "data," counts as "data"
        if len(word) >= min_length:
            words.append(word)
    print(words)
    print("Count:", len(words))

sentence = "If we have data, let us look at data. If all we have are opinions, let us go with mine"
count_words(sentence)                 
count_words(sentence, 2)              
count_words(sentence, min_length=4)   
````

## Common Mistakes: Parameter Mismatch

### Parameter mismatch — wrong number of arguments


````python
def add(a, b):
    print(a + b)

add(3, 4, 5)   # too many arguments - this will error
````

````python
def add(a, b):
    print(a + b)

add(3)   # too few arguments - this will error
````

### Mixing up parameter order with positional arguments


````python
def describe_pet(name, animal_type):
    print(f"{name} is a {animal_type}")

describe_pet("dog", "Rex")   # oops - swapped! prints "dog is a Rex"
describe_pet("Rex", "dog")   # correct order
````

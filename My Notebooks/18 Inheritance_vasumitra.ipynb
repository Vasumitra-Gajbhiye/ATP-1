# Inheritance

The Child class is called a derived class or subclass. The Parent class is known as base class or superclass. Parent is specified in () after the class name, class Child(Parent).

With inheritance, you are taking an existing class and:

- Adding new methods
- Redefining some of the existing methods
- Adding new attributes to instances
In the end you are extending existing code.

````python
class User: # Parent

    def login(self):
        print('login')

    def register(self):
        print('Register')
````

````python
class Student(User): # Child

    def enroll(self):
        print('Enroll')

    def review(self):
        print('Review')
````

````python
stu1 = Student()

stu1.enroll()
stu1.review()
stu1.login()
stu1.register()
````

````python
u = User()

u.login()
u.register()
````

````python
u.enroll()
u.review()
````

# Constructor

- `class B` inherits `class A`

- No constructor in `B` ---> `A`'s constructor invoked

- Parent class constructor used if child lacks one

````python
class Stock:
    def __init__(self, name, shares, price):
        self.name = name
        self.shares = shares
        self.price = price
        print ("Stock Constructor: ", self.name)

    def cost(self):
        return self.shares * self.price

    def sell(self, nshares):
        self.shares -= nshares
````

````python
class MyStock(Stock):
    pass
````

````python
s = MyStock('GOOGLE', 100, 490.1)
````

````python
s.sell(25)
````

````python
s.shares
````

## Method Overriding
- Child Class method overrides Parent Class method if they have the same name.

````python
class Stock:
    def __init__(self, name, shares, price):
        self.name = name
        self.shares = shares
        self.price = price

    def cost(self):
        return self.shares * self.price

    def sell(self, nshares):
        self.shares -= nshares
````

````python
class MyStock(Stock):
    def cost(self):
        return 1.25 * self.shares * self.price
````

````python
s = MyStock('GOOG', 100, 400)

print(s.cost())
````

# You can add new methods in child class

````python
class Stock:
    def __init__(self, name, shares, price):
        self.name = name
        self.shares = shares
        self.price = price

    def cost(self):
        return self.shares * self.price

    def sell(self, nshares):
        self.shares -= nshares
````

````python
class MyStock(Stock):
    def panic(self):
        self.sell(self.shares)
````

````python
s = MyStock('GOOG', 100, 1000)
s.sell(25)
print(s.shares)
````

````python
s.panic()
print(s.shares)
````

# Overriding
Sometimes a class extends an existing method, but it wants to use the original implementation inside the redefinition. For this, use super():

````python
class Stock:
    def __init__(self, name, shares, price):
        self.name = name
        self.shares = shares
        self.price = price

    def cost(self):
        return self.shares * self.price

    def sell(self, nshares):
        self.shares -= nshares
````

````python
class MyStock(Stock):
    def cost(self):
        actual_cost = super().cost()
        print("actual cost from parent is: ", actual_cost)
        return 1.25 * actual_cost
````

````python
s = MyStock('GOOG', 100, 1000)
````

````python
s.cost()
````

class Parent:
    def __init__(self, num):
        self.__num = num

    def get_num(self):

```

class Child(Parent):

```

````python
class Parent:
    def __init__(self, num):
        self.__num = num

    def get_num(self):
        return self.__num

class Child(Parent):
    def __init__(self, val, num):
        self.__val = val    # Parent's constructor is never called, so __num is never created

    def get_val(self):
        return self.__val

example1 = Child(100, 10)

print("Parent: Num:", example1.get_num())   # AttributeError: 'Child' object has no attribute '_Parent__num'
print("Child: Val:", example1.get_val())    # never reached
````

**Prediction:** `AttributeError: 'Child' object has no attribute '_Parent__num'`

- `Child` has its own `__init__`, so the parent's `__init__` is **not** run automatically.
- So `self.__num` (stored as `_Parent__num` through name mangling) is never created.
- `get_num()` is inherited and looks for `_Parent__num`. It isn't there, so Python raises the error on the first `print`, and the second `print` never runs.

**Fix:** call the parent constructor with `super().__init__(num)` inside `Child.__init__`.

````python
# Fixed version: call the Parent constructor with super()
class Child(Parent):
    def __init__(self, val, num):
        super().__init__(num)   # creates _Parent__num
        self.__val = val

    def get_val(self):
        return self.__val

example1 = Child(100, 10)

print("Parent: Num:", example1.get_num())   # Parent: Num: 10
print("Child: Val:", example1.get_val())    # Child: Val: 100
````

## Super Keyword
- Accesses parent class methods
- Accesses parent class constructor
- Does NOT access parent class attributes

````python
class Phone:
    def __init__(self, price, brand, camera):
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")
````

````python
class SmartPhone(Phone):
    def buy(self):
        print ("Buying a smartphone")
        super().buy()
````

````python
s = SmartPhone(20000, "Apple", 13)
````

````python
s.buy()
````

````python
s.super().buy() # super keyword will not work outside of the class
````

````python
class Parent:
    def __init__(self):
        self.variable1 = 100

    def display(self, variable1):
        print("Parent class: ", self.variable1)

class child(Parent):
    def display_new(self):
        print("Inside Child class: ")
````

````python
example1 = child()      # child has no __init__, so Parent's __init__ runs -> variable1 = 100

example1.display(200)   # Parent class:  100
````

**Prediction:** `Parent class:  100`

- `child` has no constructor, so `Parent.__init__` runs and sets `self.variable1 = 100`.
- `display` is inherited from `Parent`.
- `200` goes into the **local parameter** `variable1`, but the method prints `self.variable1` (the instance attribute). The argument is never used, so the output is `100`, not `200`.

````python
# The child's own method works too
example1.display_new()   # Inside Child class:

# A Parent object cannot use the child's method
p = Parent()
p.display_new()          # AttributeError: 'Parent' object has no attribute 'display_new'
````

# Multilevel level Inheritence

````python
class Product:
    def review(self):
        print ("Product customer review")

class Phone(Product):
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera
    def buy(self):
        print ("Buying a phone")

class SmartPhone(Phone):
    def __init__(self, price, brand, camera, os, ram):
        print ("Inside smartphone constructor")
        super().__init__(price, brand, camera)
        self.os = os
        self.ram = ram
    def buy(self):
        print ("Buying a smartphone")
````

````python
s = SmartPhone(20000, "Apple", 12, "IOS", 16)
````

````python
s.brand
````

````python
s.os
````

````python
p = Phone(1000, "Samsung", 1)
````

````python
s.review()
````

````python
p.review()
````

## Stack Concept — LIFO, Push/Pop

A **stack** stores items so that the **last one added is the first one removed** — LIFO: **L**ast **I**n, **F**irst **O**ut.

**Examples:**
- Your browser's **Back** button — the most recently visited page is the first one you go back to.
- **Undo** in a text editor — the most recent action is the first one undone.

**Core operations:**
| Operation | Meaning |
|---|---|
| `push` | add an item to the top |
| `pop` | remove and return the item from the top |
| `peek` | look at the top item without removing it |
| `is_empty` | check if the stack has anything in it |

## Why a linked list fits for Stacks?

### Which end of a linked list should be the "top"?

A singly linked list (with just a `head` pointer) costs:

| Operation | Cost |
|---|---|
| Add at the **head** | O(1), just re-point two arrows |
| Remove the **head** | O(1), just move `head` forward |

A stack only ever touches **one end**. If we make the **head** the top of the stack, then `push` and `pop` are both **O(1)**, with no walking and no shifting.

```
push 10, push 20, push 30 gives:

   head (TOP)
     |
     v
   [30] -> [20] -> [10] -> None
```

## Build the Stack class

````python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
````

````python
class Stack:
    def __init__(self):
        self.head = None        # head is the TOP of the stack
        self._size = 0

    def is_empty(self):
        return self.head is None

    def size(self):
        return self._size

    def __str__(self):
        items = []
        current = self.head
        while current is not None:
            items.append(str(current.data))
            current = current.next
        return "TOP -> " + " -> ".join(items) if items else "(empty stack)"

    def push(self, item):
        node = Node(item)
        node.next = self.head   # new node points at the old top
        self.head = node        # head now points at the new node
        self._size += 1

    def pop(self):
        if self.is_empty():
            return None         # nothing to pop
        item = self.head.data
        self.head = self.head.next   # move head down one node
        self._size -= 1
        return item

    def peek(self):
        if self.is_empty():
            return None
        return self.head.data
````

```python
def insertion_sort(data):

    # Loop through all data with index i
    for i in range(1, len(data)):
        print("pass", i)

    # set index j and go in reverse to check and swap
    for j in range(i, 0, -1):
        print("j", j)
        if data[j] < data[j-1]:
            data[j], data[j-1] = data[j-1], data[j]
        else:
            break

    return data

```

```python
data = [23,12,34,11,100,56,78]
insertion_sort(data)
```

```python
data = [11, 12, 23, 34, 56, 78, 100]
insertion_sort(data)
```

# Merge Sort


```python
def merge_data(data1,data2):

  i = j = 0
  result = []

  while i < len(data1) and j < len(data2):
    if data1[i] < data2[j]:
      result.append(data1[i])
      i += 1
    else:
      result.append(data2[j])
      j += 1

  return result
```

```python
data1 = [3, 4, 8]
data2 = [1, 5, 6]

merge_data(data1,data2)
```

```python
def merge_data(data1,data2):

  i = j = 0
  result = []

  while i < len(data1) and j < len(data2):
    if data1[i] < data2[j]:
      result.append(data1[i])
      i += 1
    else:
      result.append(data2[j])
      j += 1

  while i < len(data1):
    result.append(data1[i])
    i += 1

  while j < len(data2):
    result.append(data2[j])
    j += 1

  return result
```

```python
data1 = [3, 4, 8]
data2 = [1, 5, 6]

merge_data(data1,data2)
```

```python
def merge_sort(data):
  # If we have only 1 item in list then its sorted.
  if len(data) == 1:
    return data

  mid = len(data)//2

  left = data[:mid]
  right = data[mid:]

  left = merge_sort(left)
  right = merge_sort(right)

  return merge_data(left,right)
```

```python
data = [2, 1, 5, 8, 9, 6, 7, 4]

merge_sort(data)
```

```python
data = [1, 2, 4, 5, 6, 7, 8, 9]

merge_sort(data)
```

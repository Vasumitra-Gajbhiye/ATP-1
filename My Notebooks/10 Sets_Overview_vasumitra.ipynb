## Sets — Creation, Uniqueness, Set Operations

A **set** is an unordered collection of **unique** items, written with curly braces `{}` (like a dictionary, but without key-value pairs).

```python
fruits = {"apple", "banana", "cherry"}
```

Key properties:
- **No duplicates** — a set automatically removes them
- **Unordered** — items have no fixed position (no indexing!)
- Very fast to check membership (`in`) and remove duplicates

````python
fruits = {"apple", "banana", "cherry", "apple", "banana"}
print(fruits)          # duplicates are gone automatically
print(type(fruits))
print(len(fruits))
````

````python
# Creating a set from a list — a classic way to remove duplicates
numbers = [1, 2, 2, 3, 4, 4, 4, 5]
unique_numbers = set(numbers)
print(unique_numbers)
````

````python
#Create empty list and print the variable type
empty_list = []
print(type(empty_list))

#Create empty tuple and print the variable type
empty_tuple = ()
print(type(empty_tuple))

#Create empty dict and print the variable type
empty_dict = {}
print(type(empty_dict))

#Create empty set and print the variable type
empty_set = set()      # {} would give a dict, so we must use set()
print(type(empty_set))
````

````python
# Empty set — careful! {} creates an empty DICT, not a set
empty_set = set()
print(type(empty_set))

wrong_set = {}
print(type(wrong_set))   # this is a dict, not a set!
````

### Delete data from set

### Set operations — union, intersection, difference

| Operation | Symbol | Method | Meaning |
|---|---|---|---|
| Union | `\|` | `.union()` | all items from both sets |
| Intersection | `&` | `.intersection()` | items in **both** sets |
| Difference | `-` | `.difference()` | items in the first set but **not** the second |
| Symmetric difference | `^` | `.symmetric_difference()` | items in either set, but **not both** |


````python
students_python = {"Aditi", "Rohan", "Meera", "Kabir"}
students_java = {"Meera", "Kabir", "Sara", "Vikram"}

print("All students - Union:", students_python | students_java)
print("Students who both Python and Java - Intersection:", students_python & students_java)
print("Students who only know Python:", students_python - students_java)
print("Students who only know Java:", students_java - students_python)
print("Students eith know Java or Python - Symmetric difference:", students_python ^ students_java)
````

````python
students_python = {"Aditi", "Rohan", "Meera", "Kabir"}
students_java = {"Meera", "Kabir", "Sara", "Vikram"}

print("Union:", students_python.union(students_java))
print("Intersection:", students_python.intersection(students_java))
print("Python only:", students_python.difference(students_java))
print("Java only:", students_java.difference(students_python))
print("Symmetric difference:", students_python.symmetric_difference(students_java))
````

#### isdisjoint() function
checks whether two sets have any elements in common.
- It returns a boolean value:True: If the sets are completely disjoint (meaning they share absolutely no elements).
- False: If the sets share at least one common element.

````python
s1 = {1, 2, 3, 4, 5}
s2 = {4, 5, 6, 7, 8}

s1.isdisjoint(s2)
````

## Compare isdisjoint() vs intersection()

| isdisjoint() | intersection() |
|---|---|
| Primary Purpose Checks a condition (Are the sets entirely separate?) |  Finds common items (What do the sets share?) |
| Return TypeBoolean (True or False) |  A new set containing the overlapping items |
| Memory Efficiency good - because It stops scanning the moment it finds one match |  It must scan both collections fully to build the new set|

## When to Use a Set vs a List

| | List | Set |
|---|---|---|
| Ordered? | Yes | No |
| Allows duplicates? | Yes | No |
| Indexable (`lst[0]`)? | Yes | **No** |
| Membership check (`in`) | Slow for large data | Very fast |
| Best for | ordered sequences, data with duplicates, position matters | uniqueness, fast lookups, set math (union/intersection) |

**Note: ** An ordered collection like a list preserves the exact sequence in which items are added, whereas an unordered collection like a set does not track or guarantee any specific sequence

### Activity — Build it: find common interests

Ask the user (using `input()`) to type their hobbies separated by commas (e.g., `"reading, gaming, cooking"`). Split the input into a list, then convert it to a set. Compare it against the set below to find shared hobbies.

````python
friend_hobbies = {"reading", "gaming", "hiking", "painting"}

user_input = input("Enter your hobbies, separated by commas: ")

# Convert to a set of hobbies
hobby_list = user_input.split(",")
user_hobbies = set()
for hobby in hobby_list:
    hobby = hobby.strip().lower()   # remove extra spaces and ignore case
    if hobby != "":
        user_hobbies.add(hobby)
print("Your hobbies:", user_hobbies)

# Find shared hobbies with friend_hobbies
shared = user_hobbies & friend_hobbies
if shared:
    print("Shared hobbies:", shared)
else:
    print("No shared hobbies.")
````

## Decision Framework — List vs Tuple vs Dict vs Set

Use this framework when deciding which structure to reach for:

```
Do you need labeled fields (name -> value)?
    YES -> Dictionary

Do you need to guarantee uniqueness, or do fast membership checks / set?
    YES -> Set

Will the data ever change after creation (add/remove/update)?
    NO  -> Tuple
    YES -> List (unless a Set fits better, per above)
```

| Structure | Ordered | Duplicates | Mutable | Indexed by | Typical use |
|---|---|---|---|---|---|
| **List** | Yes | Yes | Yes | position | sequences that change, order matters |
| **Tuple** | Yes | Yes | No | position | fixed records, function returns |
| **Dict** | Yes (3.7+) | Keys: No | Yes | key | labeled records, lookups by name |
| **Set** | No | No | Yes | — | uniqueness, membership tests, set math |

### Activity

**Scenario:** You're tracking which students have submitted their assignment. You need to be able to quickly check if a specific student has submitted, add new submissions as they come in. Note - a student can't "submit twice" in your tracking.

Build the appropriate structure, add at least 10 student names to it, then check if any 2 students (randomly) have submitted.


````python
import random

# A set fits: fast "has X submitted?" checks, easy to add to, and no duplicate submissions
submitted = set()

submitted.add("Aditi")
submitted.add("Rohan")
submitted.add("Meera")
submitted.add("Kabir")
submitted.add("Sara")
submitted.add("Vikram")
submitted.add("Ananya")
submitted.add("Arjun")
submitted.add("Priya")
submitted.add("Dev")
submitted.add("Rohan")   # submitting twice changes nothing

print("Submitted:", submitted)
print("Number of submissions:", len(submitted))

# The full class also has some students who have not submitted yet
class_list = list(submitted) + ["Ishaan", "Neha", "Tara"]

# Pick 2 random students and check whether they submitted
for student in random.sample(class_list, 2):
    if student in submitted:
        print(student, "has submitted.")
    else:
        print(student, "has NOT submitted.")
````

## Pick the Right Structure for these Scenarios

For each scenario below:
1. State which structure you'd use (list / tuple / dict / set) and why (one sentence)
2. Write a small working code snippet that creates the structure and demonstrates ONE realistic operation on it

**Scenario A:** A shopping cart where items can be added, removed, and the same item can appear more than once (e.g., "2 apples" means "apple" appears twice).

**Scenario B:** A record of a book: its title, author, and ISBN number, which won't change once the book is catalogued.

**Scenario C:** All unique visitors (by user ID) to a webpage today — you just need to know who visited, not how many times.

**Scenario D:** A mapping of country codes to country names (e.g., `"IN"` -> `"India"`) used to look up a country quickly by its code.

**Scenario E:** A leaderboard of the top 5 players and their scores, where order (rank) matters and scores can be updated.

````python
# Scenario A
# Structure chosen: List, because items can be added or removed and the same item can appear more than once.
cart = ["apple", "bread", "apple", "milk"]
cart.append("eggs")
cart.remove("bread")
print("Cart:", cart)
print("Number of apples:", cart.count("apple"))
````

````python
# Scenario B
# Structure chosen: Tuple, because the book record is a fixed group of values that never changes once catalogued.
book = ("The Alchemist", "Paulo Coelho", "978-0062315007")
title, author, isbn = book    # unpack the record
print(f"{title} by {author} (ISBN: {isbn})")
````

````python
# Scenario C
# Structure chosen: Set, because we only need each visitor once and repeat visits should not be counted again.
visitors = set()
for user_id in ["u101", "u205", "u101", "u330", "u205", "u101"]:
    visitors.add(user_id)
print("Unique visitors:", visitors)
print("Unique visitor count:", len(visitors))
print("Did u330 visit?", "u330" in visitors)
````

````python
# Scenario D
# Structure chosen: Dictionary, because each code (key) maps to a name (value), which gives a fast lookup by code.
countries = {"IN": "India", "US": "United States", "JP": "Japan", "FR": "France"}
print(countries["IN"])
print(countries.get("BR", "Unknown code"))   # safe lookup for a missing code
````

````python
# Scenario E
# Structure chosen: List (of [name, score] pairs), because rank order matters and scores must be updatable.
leaderboard = [["Aditi", 950], ["Rohan", 900], ["Meera", 870], ["Kabir", 820], ["Sara", 800]]

# Kabir scores more points, so update Kabir's score and re-rank
for player in leaderboard:
    if player[0] == "Kabir":
        player[1] = 960

leaderboard.sort(key=lambda p: p[1], reverse=True)

for rank, (name, score) in enumerate(leaderboard, start=1):
    print(rank, name, score)
````

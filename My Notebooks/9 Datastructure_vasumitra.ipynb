Question 4 — Check stock

The warehouse considers a product to need restocking when fewer than 20 units are available. Go through all products and display products which require restocking.

````python
RESTOCK_THRESHOLD = 20

print("Products that need restocking:")
for product_id, details in products.items():
    if details["quantity"] < RESTOCK_THRESHOLD:
        print(f"{product_id} - {details['name']} - only {details['quantity']} units left")
````

Question 5 — Calculate inventory value

For every product, calculate its inventory value:
```
price × quantity
```
Display the inventory value for each product. Then calculate and display the total value of all products currently in the warehouse.

````python
total_value = 0

for product_id, details in products.items():
    inventory_value = details["price"] * details["quantity"]
    total_value += inventory_value
    print(f"{product_id} - {details['name']}: ₹{details['price']} x {details['quantity']} = ₹{inventory_value}")

print(f"Total inventory value: ₹{total_value}")
````

Question 8 — Check stock for processing a customer order

A customer provides:
- Product ID
- Quantity required

Check:
- Display `Product not found` if the Product ID does not exist.
- Display `Insufficient stock` if the requested quantity is greater than the available quantity.
- Otherwise display `Order accepted`.

````python
def check_order(product_id, quantity_required):
    product_id = product_id.strip().upper()
    if product_id not in products:
        return "Product not found"
    if quantity_required > products[product_id]["quantity"]:
        return "Insufficient stock"
    return "Order accepted"

product_id = input("Enter the Product ID: ")
quantity_required = int(input("Enter the quantity required: "))
print(check_order(product_id, quantity_required))
````

````python
# Testing all three outcomes
print("P999, 5   ->", check_order("P999", 5))    # No ID
print("P101, 20  ->", check_order("P101", 20))   # Only 15 keyboards in stock
print("P104, 100 ->", check_order("P104", 100))  # Exactly the available quantity
print("P102, 10  ->", check_order("P102", 10))
````

```python
%%writefile /content/drive/MyDrive/DemoClasses/atmclass.py

class Atm:

    def __init__(self):
        self.pin = ""
        self.balance = 0

        self.menu()

    #Getters and setters: the controlled "doors" to the private data
    def get_balance(self):
        return self.balance

    def set_balance(self, amount):
        # Setter with validation: the balance can never be negative
        if amount < 0:
            print("Balance cannot be negative")
            return False
        self.balance = amount
        return True

    def menu(self):
        # Keep showing the menu until the user chooses to exit
        while True:
            user_input = input("""
          Hi how can I help you?
          1. Press 1 to create pin
          2. Press 2 to deposit
          3. Press 3 to withdraw
          4. Press 4 to check balance
          5. Anything else to exit""")

            if user_input == "1":
                self.create_pin()
            elif user_input == "2":
                self.deposit()
            elif user_input == "3":
                self.withdraw()
            elif user_input == "4":
                self.check_balance()
            else:
                print("Thank you for transacting with us.")
                break

    def create_pin(self):
        self.pin = input("Enter your pin: ")
        print("Pin set successfully")

    def deposit(self):
        temp = input("Enter your pin: ")
        if temp == self.pin:
            amount = int(input("Enter the amount: "))
            if amount <= 0:
                print("Amount must be positive")
            elif self.set_balance(self.get_balance() + amount):
                print("Deposit successful")
        else:
            print("Invalid pin")

    def withdraw(self):
        temp = input("Enter your pin: ")
        if temp == self.pin:
            amount = int(input("Enter the amount: "))
            if amount <= 0:
                print("Amount must be positive")
            elif amount <= self.get_balance():
                self.set_balance(self.get_balance() - amount)
                print("Operation successful")
            else:
                print("insufficient funds")
        else:
            print("invalid pin")

    def check_balance(self):
        temp = input("Enter your pin: ")
        if temp == self.pin:
            print("Your balance is:", self.get_balance())
        else:
            print("invalid pin")
```

## Strings:
The string values that literally appear in our code are called string literals.
- String literals can begin and end with double quotes as well as with single quotes. For Multiline Strings triple quotes are used.
- An escape sequence lets you use characters that are otherwise impossible to put into a string literal. For example, \' is the escape sequence for a single quote and \n is the escape sequence for a newline character.
Ex: ```spam = 'Say hi to Bob\'s mother.'```

| Escape sequence | Prints as ... |
|---|---|
| \' | Single quote |
| \" | Double quote |
| \t | Tab |
| \n | Newline (line break) |
| \\ | Backslash |

##Numeric Code Points of Characters
Computers store information as bytes, which means the text can be converted to numbers. Because of this requirement, every text character has a corresponding numeric value called a Unicode code point. For example, the numeric code point is 65 for 'A', 52 for '4', and 33 for '!'. You can use the ord() function to get the code point of a one-character string, and the chr() function to get the one-character string of an integer code point.

````python
print(ord('A'))

print(ord('4'))

print(ord('!'))

print(chr(65))
````

## String Basics & Indexing

A string is a sequence of characters. Each character has a **position (index)**, starting from **0**.

```
 P   Y   T   H   O   N
 0   1   2   3   4   5
-6  -5  -4  -3  -2  -1
```

Notice: Python also supports **negative indexing**, counting backwards from -1 (the last character).


````python
word = "PYTHON"
print(word[0])    # first character
print(word[3])    # 4th character
print(word[-1])   # last character
print(word[-2])   # second-to-last character
print(len(word))  # length of the string
````

## Slicing

Slicing extracts a **substring** using the syntax:

```python
string[start:stop:step]
```

- `start` — index to begin at (included), default is 0
- `stop` — index to stop before (**excluded**), default is end of string
- `step` — how many characters to skip, default is 1

Any of these can be left out.

````python
word = "PROGRAMMING"

print(word[0:4])     # characters at index 0,1,2,3
print(word[4:])      # from index 4 to the end
print(word[:4])      # from the start up to (not including) index 4
print(word[:])       # the whole string (a copy)
print(word[::2])     # every 2nd character
print(word[::-1])    # the whole string, reversed!
````

## Common String Methods

Strings have many built-in methods. A method is called using `string.method()`.

| Method | What it does |
|---|---|
| `.upper()` | converts to uppercase |
| `.lower()` | converts to lowercase |
| `.strip()` | removes leading/trailing whitespace |
| `.split(sep)` | splits into a list of substrings |
| `.replace(old, new)` | replaces all occurrences |
| `.find(sub)` | returns index of first occurrence (-1 if not found) |
| `.count(sub)` | counts occurrences |
| `.startswith(sub)` / `.endswith(sub)` | checks the start/end |
| `.join(list)` | joins a list of strings using this string as separator |
| `Capitalize/Title/Upper/Lower/Swapcase` | Change string case |
| `isalnum/isalpha/isdecimal/isdigit/isidentifier` | Check the datatype |


Strings are **immutable** — these methods return a *new* string, they don't change the original.


````python
text = "  Hello, Python World!  "

print(text.strip())
print(text.replace("Python", "Coding"))
print(text.find("Python"))
print(text.count("o"))
print(text.strip().startswith("Hello"))
print(text.strip().endswith("!"))
````

````python
print('it is raining today'.capitalize())
print('it is raining today'.title())
print('it is raining today'.lower())
print('it is raining today'.upper())
print('it is raining today'.swapcase())
````

````python
print("FLAT204".isalnum()) # Alphanumeric
print("FLAT20&".isalnum())
print("FLAT".isalpha()) # Alphabetic
print("20".isdigit())
print("Hello World".isidentifier())
print("Hello_World".isidentifier())
````

**Recap: What is an Identifier?**
- An identifier is the term for a name we give to a variable name, a function name, a class name, etc.

- .isidentifier() is a string method that checks: "could this text legally be used as a name in Python code?"

- It returns True or False

- Note: It does not check whether the name already exists or is in use, only whether it could be used as one.

````python
sentence = "the quick brown fox jumps"
words = sentence.split()          # splits on whitespace by default
print(words)
print(len(words))

fruits_line = "apple,banana,cherry"
fruits = fruits_line.split(",")      # split on a specific character
print(fruits)

joined = "-".join(fruits)         # opposite of split
print(joined)
````

## **Question:**
**String is Immutable and hence we saw del and index assignment did not work. But then how a .replace() works?**

- del and index assignment, are asking Python to reach into the existing string object in memory and physically change one of its characters, or remove one. This is violation of Immutabaility. Hence the error.
- `.replace()` doesn't touch the original at all — it builds a brand new string

**Note: Every single string method — .replace(), .upper(), .strip(), all of them — returns a new string and leaves the original untouched.**

````python
## You can check the actual memory identity using id(), which shows you the memory address of an object

s = "hello"
print(id(s))

s = s.replace("h", "H")
print(id(s))
````

- range(n) - Creates a sequence of numbers from 0 to n-1
- range(a, b) - Creates a sequence of numbers from a to b-1

## For Loops

A `for` loop repeats a block of code **once for each item** in a sequence.

```python
for variable in sequence:
    # code that runs each time
```

The most common sequence to loop over is `range(...)`, which generates a sequence of numbers.


````python
for i in range(5):
    print(i)
````

````python
print("range(5):")
for i in range(5):
    print(i, end=" ")

print("\n\nrange(2, 8):")
for i in range(2, 8):
    print(i, end=" ")

print("\n\nrange(0, 10, 2):")
for i in range(0, 10, 2):
    print(i, end=" ")
````

## While Loops

A `while` loop repeats **as long as a condition is `True`**. Unlike `for`, you don't know in advance how many times it will run.

```python
while condition:
    # code that runs while condition is True
    # something must eventually make condition False!
```

````python
count = 0
while count < 5:
    print(count)
    count += 1     # important! without this, the loop never ends
````

## Loop Control: `break` and `continue`

- **`break`** — exits the loop immediately, skipping any remaining iterations.
- **`continue`** — skips the rest of the current iteration and moves to the next one.


````python
for i in range(10):
    if i == 5:
        break
    print(i)
print("Loop ended with break")
````

````python
for i in range(10):
    if i % 2 == 0:
        continue
    print(i)
print("Loop ended with continue — only odd numbers printed")
````

### Infinite Loops — a common pitfall

An infinite loop happens when the condition in a `while` loop never becomes `False`. This freezes your program!

```python
# DANGER — do not run this as-is, it never stops:
# n = 1
# while n > 0:
#     print(n)
```

### Activity - Use loop to iterate from 0 to n (n provided by user) and print the sum of all evens and the sum of all odds.

````python
# Code here

n = int(input("Enter a number: "))
sum_of_evens = 0
sum_of_odds = 0
for i in range(n + 1):
    if i % 2 == 0:
        sum_of_evens += i
    else:
        sum_of_odds += i
print("Sum of evens:", sum_of_evens)
print("Sum of odds:", sum_of_odds)



````

### Activity: Number guessing game

````python
import random
````

````python
# Create Guess the number game
jackpot = random.randint(1, 100)
guess = int(input("Provide your guess: "))
counter = 1

while guess != jackpot:
    if guess < jackpot:
        print("Guess higher")
    else:
        print("Guess lower")
    guess = int(input("Provide your guess: "))
    counter += 1

print("Your guess is correct")
print("You took", counter, "attempts")
````

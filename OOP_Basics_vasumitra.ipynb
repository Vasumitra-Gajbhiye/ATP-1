So far we've stored related data using dictionaries, tuples aor any other datatypes, and behaviour using functions — but they were separate. **Object-Oriented Programming (OOP)** lets us bundle **data (attributes)** and **behaviour (methods)** together into one thing: an **object**.

## What is a class?
Class is the Blueprint

Ex: `User Class`
- "What it takes"  -->  These are the **Attributes** (the data variables like username and password).
- "What it checks for and what message it gives"  --> These are the **Methods**.

### Ex: Cookie Cutter and Cookies

- A **class** is like a **cookie cutter** — a template/blueprint that defines the *shape* every cookie will have.
- An **object** is an actual **cookie** made using that cutter — a specific instance with its own values.

You can make many cookies (objects) from one cutter (class), and each cookie can be decorated differently (have different attribute values), but they all share the same basic shape (structure).

Class:
 - Attributes: Data/Properties
 - Methods: Functions/Behavior

## Implementation of Class and Object

### Lets Create a Class

Example: Let us create a class ATM
Now what all attributes I need?
So what behaviours I need to encode?

```python
class Atm:
    # we always define data within a sepcial method called __init__
    # __init__ is a constructor
    # Special method inside a class.
    # Initializes instance members.
    # Executes automatically on object creation.
    def __init__(self):
        self.pin = ""
        self.balance = 0
```

```python
# Creating an object (instance) of the class - __init__ runs automatically
sbi = Atm()

print(type(sbi))
print("pin:", repr(sbi.pin))      # attributes are accessed with object.attribute
print("balance:", sbi.balance)
```

```python
# One class (cookie cutter) -> many independent objects (cookies)
hdfc = Atm()

sbi.balance = 5000
print("sbi balance:", sbi.balance)    # 5000
print("hdfc balance:", hdfc.balance)  # 0 -> each object has its own copy of the attributes
```

### Adding Methods (Behaviour)

Now the behaviours our ATM needs:
- `menu` — show the options and call the right method
- `create_pin` — set a pin
- `deposit` — add money (after checking the pin)
- `withdraw` — take money out (after checking the pin and the balance)
- `check_balance` — show the balance (after checking the pin)

Every method takes `self` as its first parameter. `self` is the object the method is called on, so `self.pin` and `self.balance` are *that* object's attributes.

```python
class Atm:
    def __init__(self):
        self.pin = ""
        self.balance = 0

    def menu(self):
        while True:
            user_input = input("""
          Hi how can I help you?
          1. Press 1 to create pin
          2. Press 2 to deposit
          3. Press 3 to withdraw
          4. Press 4 to check balance
          5. Anything else to exit
          """)

            if user_input == "1":
                self.create_pin()
            elif user_input == "2":
                self.deposit()
            elif user_input == "3":
                self.withdraw()
            elif user_input == "4":
                self.check_balance()
            else:
                print("Thank you for transacting with us.")
                break

    def create_pin(self):
        self.pin = input("Enter your pin: ")
        print("Pin set successfully")

    def deposit(self):
        temp = input("Enter your pin: ")
        if temp == self.pin:
            amount = int(input("Enter the amount: "))
            self.balance = self.balance + amount
            print("Deposit successful")
        else:
            print("Invalid pin")

    def withdraw(self):
        temp = input("Enter your pin: ")
        if temp == self.pin:
            amount = int(input("Enter the amount: "))
            if amount <= self.balance:
                self.balance = self.balance - amount
                print("Operation successful")
            else:
                print("insufficient funds")
        else:
            print("invalid pin")

    def check_balance(self):
        temp = input("Enter your pin: ")
        if temp == self.pin:
            print(self.balance)
        else:
            print("invalid pin")
```

### Calling Methods on an Object

Create an object and call `menu()` on it. Try: `1` (create pin) → `2` (deposit) → `3` (withdraw) → `4` (check balance) → any other key to exit.

```python
sbi = Atm()
sbi.menu()
```

Methods can also be called one at a time. `sbi.deposit()` is just Python's shorthand for `Atm.deposit(sbi)` — the object before the dot is passed in as `self`.

```python
hdfc = Atm()

hdfc.create_pin()
hdfc.deposit()
Atm.check_balance(hdfc)   # same as hdfc.check_balance()
```
